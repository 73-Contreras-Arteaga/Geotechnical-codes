1.1. Modelos de caracterización de macizos rocosos y
bloques

Los modelos de caracterización de macizos rocosos y bloques son herramientas y metodologías utilizadas en la geomecánica para describir y cuantificar las propiedades mecánicas de una masa rocosa. A diferencia de una roca intacta, un macizo rocoso está compuesto por la matriz rocosa y un conjunto de discontinuidades (fallas, diaclasas, estratificación, etc.) que afectan significativamente su resistencia y deformabilidad.

Estos modelos buscan representar la complejidad del macizo rocoso de manera simplificada, permitiendo predecir su comportamiento ante excavaciones, cimentaciones o cualquier otra intervención ingenieril. Algunos aspectos clave que se consideran en estos modelos incluyen:

*   **Propiedades de la roca intacta:** Resistencia a la compresión uniaxial (UCS), resistencia a la tracción, módulo de Young, relación de Poisson.
*   **Características de las discontinuidades:** Orientación, espaciamiento, persistencia, rugosidad, relleno y resistencia al corte.
*   **Índices de calidad del macizo rocoso:** Sistemas como el RMR (Rock Mass Rating) de Bieniawski o el sistema Q de Barton, que integran diversas características para asignar un valor numérico a la calidad del macizo.
*   **Clasificaciones geomecánicas:** Como se mencionó, RMR y Q son ampliamente utilizados para clasificar los macizos rocosos y proporcionar una estimación de sus propiedades y la necesidad de sostenimiento.

El objetivo final es obtener parámetros ingenieriles que puedan ser utilizados en análisis de estabilidad y diseño de obras subterráneas o taludes, garantizando la seguridad y eficiencia de los proyectos.
