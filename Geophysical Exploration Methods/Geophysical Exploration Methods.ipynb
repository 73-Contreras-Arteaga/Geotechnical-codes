## Metodos de exploración geofisicos

# 1. Generalidades de los métodos geofísicos

## 1.1. Descripción de los métodos geofísicos.

En el vasto campo de la exploración geotécnica y la ingeniería civil, los **métodos geofísicos** representan un conjunto de herramientas indirectas no invasivas que permiten inferir propiedades del subsuelo a partir de la medición de parámetros físicos en la superficie o a poca profundidad. Estos métodos se basan en principios de la física, como la propagación de ondas sísmicas, la conductividad eléctrica, las variaciones del campo gravitatorio o las anomalías del campo magnético terrestre, para caracterizar geológicamente el terreno sin necesidad de perforaciones extensivas.

La esencia de la geofísica aplicada radica en detectar contrastes en las propiedades físicas de los materiales del subsuelo. Por ejemplo, una roca compacta tendrá una respuesta sísmica o eléctrica diferente a la de un suelo suelto o saturado de agua. Estas diferencias, una vez medidas y analizadas, se correlacionan con características ingenieriles relevantes como la litología, la presencia de fallas, la ubicación del nivel freático, la compactación del suelo, o la detección de cavidades y cuerpos intrusivos.

La aplicación de estos métodos en geotecnia es crucial para:

*   **Delimitar capas estratigráficas:** Identificar la profundidad y extensión de diferentes unidades geológicas.
*   **Caracterizar propiedades mecánicas:** Inferir parámetros como la rigidez dinámica (velocidades de ondas sísmicas) o la resistividad para evaluar la capacidad portante o la compactación.
*   **Detectar anomalías:** Localizar zonas de baja densidad (cavitaciones, rellenos poco compactados), alta conductividad (arcillas, zonas saturadas), o cambios abruptos que puedan indicar fallas o estructuras geológicas complejas.
*   **Monitoreo:** En algunos casos, se utilizan para observar cambios en el subsuelo a lo largo del tiempo, como movimientos de tierra o variaciones del nivel freático.

Es fundamental comprender que los métodos geofísicos proporcionan información indirecta. La interpretación de los datos requiere no solo un sólido conocimiento de los principios físicos, sino también una adecuada contextualización geológica y geotécnica, a menudo complementada con datos directos (sondeos, ensayos de laboratorio) para una calibración y validación robustas. La elección del método más adecuado depende de los objetivos específicos del estudio, las características geológicas esperadas y las condiciones del sitio.

## 1.2. Características y diferencias de cada método.

Dentro del arsenal de los métodos geofísicos, la elección del enfoque adecuado es primordial y depende directamente de los objetivos específicos del estudio geotécnico, la profundidad de investigación requerida, el tipo de anomalías esperadas y las condiciones del sitio. Aunque todos buscan inferir propiedades del subsuelo, sus principios físicos, la información que proporcionan y sus limitaciones difieren significativamente.

A continuación, se describen las características fundamentales y las principales diferencias entre los métodos geofísicos más comúnmente aplicados en geotecnia:

### 1. Métodos Sísmicos

*   **Principio:** Se basan en la generación y propagación de ondas elásticas (sísmicas) a través del subsuelo. La velocidad y la forma en que estas ondas viajan y se reflejan o refractan dependen de las propiedades elásticas de los materiales (densidad, módulo de elasticidad).
*   **Tipos principales:**
    *   **Refracción Sísmica:** Mide el tiempo que tardan las ondas P (compresionales) en viajar a través de diferentes capas del subsuelo, refractándose en los límites de estas capas. Es ideal para determinar la profundidad y el espesor de capas estratigráficas y para estimar propiedades geotécnicas como el módulo de Young dinámico.
    *   **Reflexión Sísmica:** Mide el tiempo que tardan las ondas en reflejarse en los límites de las capas. Permite una resolución más fina de estructuras profundas, aunque su aplicación en geotecnia superficial es menos común que la refracción.
    *   **MASW (Multichannel Analysis of Surface Waves):** Analiza la dispersión de las ondas superficiales (ondas Rayleigh) para determinar los perfiles de velocidad de ondas S (cizalla). Es excelente para la caracterización del subsuelo a poca profundidad, la identificación de zonas blandas, la estimación del $V_S$ (velocidad de onda de cizalla), y la clasificación sísmica de sitios (parámetros como $V_{S30}$).
*   **Información clave:** Profundidad de capas, espesores, velocidades de ondas P y S, módulos de rigidez dinámicos, identificación de roca madre, zonas fracturadas.
*   **Ventajas:** Alta resolución para interfaces bien definidas, información directa sobre rigidez, aplicable en diversos entornos.
*   **Limitaciones:** Requiere una fuente de energía (martillo, peso, explosivo), susceptible al ruido ambiental, interpretación compleja en zonas de velocidades invertidas.

### 2. Métodos Eléctricos

*   **Principio:** Miden la resistividad eléctrica o la conductividad del subsuelo. Estas propiedades están intrínsecamente ligadas al contenido de humedad, la porosidad, la salinidad del agua intersticial, la presencia de arcillas y la litología de los materiales.
*   **Tipos principales:**
    *   **Sondeos Eléctricos Verticales (SEV):** Implican la inyección de corriente en el subsuelo y la medición de la diferencia de potencial, variando la separación de los electrodos para investigar diferentes profundidades. Útil para determinar la resistividad de capas horizontales y la profundidad del nivel freático.
    *   **Tomografía Eléctrica (ERT o REI):** Es una extensión 2D o 3D de los SEV, donde se utilizan múltiples electrodos para obtener una imagen de la distribución de resistividades en una sección o volumen del subsuelo. Excelente para la detección de cavidades, intrusiones, fracturas, plumas de contaminación y zonas de saturación.
*   **Información clave:** Distribución de resistividad/conductividad, profundidad del nivel freático, detección de anomalías (arcillas, arena, roca, vacíos, agua salada).
*   **Ventajas:** Equipo relativamente sencillo, alta sensibilidad a la presencia de agua y arcillas, buena resolución lateral con ERT.
*   **Limitaciones:** Sensible a interferencias electromagnéticas, la interpretación puede ser ambigua si múltiples materiales tienen resistividades similares, penetración limitada en suelos muy conductivos.

### 3. Métodos Gravimétricos

*   **Principio:** Miden las pequeñas variaciones del campo gravitatorio terrestre causadas por contrastes de densidad en el subsuelo. Las masas más densas generan una atracción gravitatoria mayor, mientras que las menos densas generan una menor.
*   **Aplicación en Geotecnia:** Principalmente para la detección de cavidades, túneles abandonados o zonas de baja densidad (karst) que podrían comprometer la estabilidad de estructuras.
*   **Información clave:** Anomalías de densidad, ubicación y geometría de cuerpos subsuperficiales con contraste de densidad.
*   **Ventajas:** No invasivo, puede detectar anomalías profundas.
*   **Limitaciones:** Requiere equipos de alta precisión, la señal puede ser sutil, la interpretación no es única (múltiples distribuciones de masa pueden generar la misma anomalía gravitatoria).

### 4. Métodos Magnéticos (Magnetometría)

*   **Principio:** Miden las variaciones en el campo magnético terrestre inducidas por la presencia de materiales con diferentes propiedades magnéticas en el subsuelo (susceptibilidad magnética).
*   **Aplicación en Geotecnia:** Útil para la detección de objetos ferrosos enterrados (tuberías, tambores, escombros metálicos), identificación de fallas geológicas (si hay mineralización asociada), y cartografía de ciertos tipos de rocas.
*   **Información clave:** Anomalías magnéticas, presencia de materiales ferromagnéticos, ciertos tipos de rocas.
*   **Ventajas:** Rápido y eficiente para grandes áreas, pasivo (no requiere inyección de energía).
*   **Limitaciones:** Sensible a interferencias artificiales (líneas eléctricas, vehículos), la interpretación es compleja y no siempre única.

### Consideraciones para la Selección del Método:

La selección de un método geofísico no es trivial y debe ser precedida por una investigación preliminar exhaustiva. Es común y a menudo recomendable utilizar una combinación de métodos (abordaje multidisciplinario) para obtener una visión más completa y reducir las ambigüedades en la interpretación de los datos del subsuelo. La complementariedad de la información obtenida por diferentes principios físicos suele conducir a una caracterización geotécnica más robusta y fiable.

## 1.3. Ventajas y desventajas de los métodos geofísicos aplicados a geotecnia.

La aplicación de los métodos geofísicos en geotecnia ha revolucionado la forma en que se realiza la exploración del subsuelo, ofreciendo una alternativa y un complemento valioso a los métodos tradicionales directos. Sin embargo, como toda herramienta de ingeniería, poseen un conjunto de ventajas y desventajas que deben ser cuidadosamente consideradas para su correcta implementación y la interpretación de sus resultados.

### Ventajas de los Métodos Geofísicos en Geotecnia:

1.  **Carácter No Invasivo y No Destructivo:** Permiten investigar el subsuelo sin necesidad de perforaciones extensivas o excavaciones, minimizando la alteración del sitio y los impactos ambientales. Esto es especialmente beneficioso en áreas urbanas o en proyectos con restricciones ambientales.
2.  **Eficiencia y Rapidez en la Adquisición de Datos:** La mayoría de los métodos geofísicos permiten cubrir grandes extensiones de terreno en un tiempo relativamente corto, proporcionando una gran cantidad de datos que pueden generar una imagen tridimensional del subsuelo.
3.  **Reducción de Costos:** Al ser más rápidos y menos intrusivos que los métodos de exploración directa (sondeos, calicatas), pueden reducir significativamente los costos de la fase de investigación, especialmente en proyectos de gran envergadura.
4.  **Información Continua y de Gran Cobertura:** A diferencia de los sondeos puntuales, los métodos geofísicos ofrecen una visión continua de las propiedades del subsuelo a lo largo de líneas o en áreas, lo que permite identificar heterogeneidades y anomalías que podrían pasarse por alto con métodos puntuales.
5.  **Detección de Anomalías y Heterogeneidades:** Son excelentes para localizar características ocultas como cavidades, zonas de baja densidad, fallas, fracturas, cuerpos de agua subterránea, plumas de contaminación, objetos enterrados o la interfaz roca-suelo.
6.  **Estimación de Propiedades Físicas y Mecánicas:** Proporcionan datos indirectos que pueden correlacionarse con propiedades geotécnicas clave, como la rigidez (velocidades de ondas sísmicas), la porosidad, el grado de saturación o la litología, lo cual es fundamental para el diseño ingenieril.
7.  **Seguridad:** Reducen la exposición del personal a riesgos asociados con la operación de equipos de perforación o excavación.
8.  **Monitoreo:** Algunos métodos pueden ser empleados para el monitoreo de cambios en el subsuelo a lo largo del tiempo, como asentamientos, infiltraciones o variaciones del nivel freático.

### Desventajas de los Métodos Geofísicos en Geotecnia:

1.  **Carácter Indirecto de la Información:** Los métodos geofísicos no proporcionan muestras directas del terreno. La información es siempre una inferencia de las propiedades físicas y requiere una interpretación experta para ser traducida a parámetros geotécnicos.
2.  **Ambigüedad en la Interpretación (No Unicidad):** Diferentes configuraciones geológicas y materiales pueden producir respuestas geofísicas similares, lo que puede llevar a interpretaciones múltiples. Por ejemplo, una zona de baja resistividad podría ser arcilla, agua salina o un suelo muy saturado.
3.  **Necesidad de Calibración con Datos Directos:** Para una interpretación fiable, es casi indispensable calibrar los resultados geofísicos con datos de exploración directa (sondeos, ensayos de laboratorio). Sin esta calibración, las incertidumbres pueden ser elevadas.
4.  **Resolución Limitada:** La capacidad de resolución de los métodos geofísicos está inversamente relacionada con la profundidad de investigación. A mayor profundidad, menor es la resolución de las anomalías o capas.
5.  **Sensibilidad a las Condiciones del Sitio:** Factores como el ruido cultural (tráfico, líneas eléctricas), las condiciones climáticas, la topografía o la vegetación pueden afectar la calidad de los datos adquiridos.
6.  **Dependencia del Contraste de Propiedades:** Los métodos geofísicos funcionan mejor cuando existen contrastes significativos en las propiedades físicas de los materiales del subsuelo. Si las propiedades son muy homogéneas, la detección de interfaces o anomalías es difícil.
7.  **Requerimientos de Equipo y Software Especializado:** La adquisición y el procesamiento de datos geofísicos requieren equipos específicos y software avanzado, así como personal capacitado en su manejo y en la interpretación.
8.  **Costo Inicial del Equipo:** Aunque a largo plazo pueden ser más económicos, la inversión inicial en equipos geofísicos puede ser considerable.

En resumen, los métodos geofísicos son herramientas extremadamente potentes y versátiles en la ingeniería geotécnica, pero su aplicación exitosa depende de una planificación cuidadosa, una adquisición de datos rigurosa, una interpretación experta y, crucialmente, de su integración con información geotécnica directa para mitigar las ambigüedades inherentes a su naturaleza indirecta.

# 2. Aplicación de métodos sísmicos en geotecnia

## 2.1. Disposición, rendimiento y limitaciones del método sísmico.

El método sísmico, en sus diversas modalidades (refracción, reflexión, MASW), constituye una herramienta fundamental en la exploración geotécnica, permitiendo caracterizar el subsuelo a través de la respuesta de materiales a la propagación de ondas elásticas. Sin embargo, su aplicación efectiva requiere comprender su disposición, evaluar su rendimiento esperado y reconocer sus inherentes limitaciones.

### Disposición del Método Sísmico

La disposición se refiere a la configuración geométrica de la fuente de energía sísmica y los geófonos (sensores) en el terreno. La elección de la disposición es crítica y depende del objetivo del estudio, la profundidad de investigación, la geología esperada y el tipo específico de método sísmico empleado.

1.  **Disposición para Refracción Sísmica (2D):**
    *   **Configuración:** Típicamente, los geófonos (generalmente de 12 a 24 o más) se colocan en línea recta a intervalos equiespaciados (ej., 2-5 m). La fuente de energía (martillo, peso, explosivo) se ubica en los extremos de la línea de geófonos (disparos *end-on*) y en puntos intermedios (disparos *off-set* o *mid-spread*) para garantizar una cobertura adecuada y redundancia en la adquisición de datos.
    *   **Objetivo:** Obtener perfiles de velocidades de ondas P (y ocasionalmente S) para la determinación de capas estratigráficas, profundidad de la roca madre, y estimación de parámetros dinámicos del suelo.
    *   **Profundidad de Investigación:** Depende de la longitud del arreglo de geófonos y la energía de la fuente. Generalmente, la profundidad máxima es aproximadamente 1/3 a 1/5 de la longitud del arreglo.

2.  **Disposición para MASW (Multichannel Analysis of Surface Waves):**
    *   **Configuración:** Similar a la refracción, con un arreglo lineal de geófonos (usualmente 24 o más) y la fuente sísmica ubicada a una distancia fija (*offset*) del primer geófono. Se realizan múltiples disparos a lo largo del arreglo o se utilizan arreglos móviles (*roll-along*) para generar perfiles 2D.
    *   **Objetivo:** Determinar el perfil de velocidades de ondas S con la profundidad ($V_S$) para la caracterización geotécnica de la parte más superficial del subsuelo, la clasificación sísmica de sitios ($V_{S30}$), y la identificación de zonas blandas.
    *   **Profundidad de Investigación:** La profundidad efectiva está relacionada con la longitud de onda de las ondas superficiales analizadas, y por ende, con la longitud del arreglo de geófonos. Puede ser comparable o ligeramente superior a la mitad de la longitud del arreglo.

3.  **Disposición para Reflexión Sísmica (común en hidrocarburos, menos en geotecnia):**
    *   **Configuración:** Geófonos y fuentes se distribuyen para captar las ondas reflejadas en las interfaces. En geotecnia superficial, se utilizan configuraciones de alta resolución con espaciados pequeños.
    *   **Objetivo:** Obtener imágenes de alta resolución de estructuras y estratos a mayor profundidad.

### Rendimiento del Método Sísmico

El rendimiento se mide por la calidad de la información obtenida (resolución, precisión) y la eficiencia operativa.

*   **Resolución:** Los métodos sísmicos tienen una buena resolución para identificar interfaces con contrastes significativos de impedancia acústica (producto de la densidad y la velocidad de onda). La resolución es inversamente proporcional a la profundidad y directamente a la frecuencia de las ondas.
*   **Precisión:** La precisión en la determinación de velocidades y profundidades depende de la calidad de los datos, la correcta aplicación de las técnicas de procesamiento y la validación con datos de perforación.
*   **Eficiencia:** La adquisición de datos puede ser relativamente rápida en condiciones óptimas, cubriendo grandes extensiones en un día. El procesamiento e interpretación, sin embargo, pueden requerir tiempo y software especializado.
*   **Información Geotécnica Directa:** Proporcionan valores de velocidad de ondas P ($V_P$) y S ($V_S$), que son parámetros fundamentales para el cálculo de módulos de elasticidad dinámicos (módulo de Young dinámico, módulo de cizalla dinámico, relación de Poisson dinámica), esenciales en análisis de respuesta sísmica de sitios, diseño de cimentaciones dinámicas y evaluaciones de compactación.

### Limitaciones del Método Sísmico

A pesar de sus ventajas, los métodos sísmicos enfrentan varias limitaciones que deben ser consideradas:

1.  **Principio de Capa de Velocidad Creciente (Refracción):** El método de refracción sísmica asume que la velocidad de las ondas aumenta con la profundidad. No puede detectar eficazmente 'capas ocultas' (capas delgadas de baja velocidad bajo una capa de alta velocidad) o 'inversiones de velocidad' (capas de menor velocidad bajo capas de mayor velocidad), lo que puede llevar a una subestimación de la profundidad de la roca madre o la omisión de estratos importantes.
2.  **Susceptibilidad al Ruido:** Las mediciones sísmicas son sensibles al ruido ambiental (tráfico, maquinaria, viento, fuentes culturales), que puede enmascarar la señal y degradar la calidad de los datos. Se requiere planificación para minimizarlo o utilizar técnicas de procesamiento avanzadas.
3.  **Resolución Limitada a Profundidad:** Como se mencionó, la resolución disminuye con la profundidad. Esto significa que capas delgadas o pequeñas anomalías pueden no ser detectables a gran profundidad.
4.  **Costo de Equipo y Expertise:** El equipo de adquisición sísmica puede ser costoso, y el procesamiento e interpretación de los datos requieren personal altamente capacitado y software especializado.
5.  **Condiciones del Terreno:** Terrenos muy blandos, suelos orgánicos o terrenos con fuertes irregularidades topográficas pueden dificultar la generación y recepción de las ondas, así como la colocación adecuada de los geófonos.
6.  **Interpretación Ambígua (especialmente en 2D):** Aunque los métodos sísmicos proporcionan información valiosa, la interpretación de las capas y sus propiedades es indirecta y requiere correlación con datos directos de perforación para su validación y para resolver posibles ambigüedades.
7.  **No Detecta Vacíos en Ciertas Condiciones:** Si un vacío o cavidad no presenta un contraste de impedancia acústica significativo con el material circundante, o si es demasiado pequeño para la longitud de onda utilizada, puede no ser detectado.

En conclusión, el método sísmico es una herramienta geofísica poderosa para la caracterización geotécnica. Su éxito radica en una adecuada planificación de la disposición, la comprensión de su rendimiento bajo las condiciones del sitio y un reconocimiento explícito de sus limitaciones, siempre buscando complementariedad con otros métodos de exploración.

## 2.2. Instalación de equipos y toma de datos en el campo.

La correcta **instalación de equipos y la toma de datos en campo** son fases críticas para asegurar la calidad y fiabilidad de los resultados obtenidos mediante métodos sísmicos. Una mala praxis en esta etapa puede invalidar el esfuerzo posterior de procesamiento e interpretación. Este proceso requiere una planificación meticulosa, el uso adecuado de la instrumentación y el seguimiento riguroso de protocolos.

### 1. Planificación Previa a la Campaña de Campo

Antes de salir al campo, es indispensable realizar una planificación detallada:

*   **Reconocimiento del sitio:** Visita preliminar para evaluar la accesibilidad, topografía, geología superficial, presencia de ruido cultural (fuentes de vibración como tráfico, fábricas) y posibles obstáculos.
*   **Diseño del levantamiento:** Definición de la ubicación y longitud de las líneas sísmicas, espaciado entre geófonos, puntos de disparo (fuente), y el tipo de equipo a utilizar, basándose en los objetivos del estudio y la geología esperada.
*   **Selección de equipos:** Elección del sismógrafo multicanal, tipo y frecuencia de geófonos (generalmente de 4.5 Hz para aplicaciones geotécnicas), fuente de energía (martillo instrumentado con placa metálica, 'weight drop', o explosivos controlados para mayores profundidades).
*   **Permisos y seguridad:** Obtención de permisos de acceso y operación, y establecimiento de protocolos de seguridad, especialmente si se utilizan fuentes de energía de alto impacto o explosivos.

### 2. Instalación del Equipo en Campo

Una vez en el sitio, la instalación sigue los siguientes pasos:

*   **Sismógrafo:** Se ubica en un lugar central y seguro, con buena visibilidad y fácil acceso a los arreglos de geófonos.
*   **Geófonos:** Se despliegan a lo largo de la línea sísmica, clavándolos firmemente en el suelo para asegurar un buen acoplamiento sísmico. El espaciado debe ser uniforme y acorde al diseño del levantamiento. Es crucial verificar que todos los geófonos estén correctamente conectados y funcionen (mediante pruebas de continuidad o monitoreo en el sismógrafo).
*   **Cables:** Los cables que conectan los geófonos al sismógrafo deben extenderse sin tensiones excesivas, evitando cruces con caminos o zonas de paso para prevenir daños.
*   **Fuente sísmica:** Se posiciona en los puntos de disparo predefinidos. Para fuentes de impacto (martillo), se utiliza una placa metálica para asegurar una transmisión eficiente de la energía al suelo. Si es instrumentado, el martillo tiene un sensor para registrar el instante exacto del impacto (tiempo cero).

### 3. Toma de Datos (Adquisición Sísmica)

La adquisición de datos es el proceso de generar las ondas sísmicas y registrarlas con el sismógrafo. Se realiza de manera sistemática para cada punto de disparo:

*   **Configuración del sismógrafo:** Se establecen los parámetros de adquisición: frecuencia de muestreo (determina la resolución temporal), duración de la ventana de registro, número de apilamientos (stacks) para mejorar la relación señal/ruido, y filtros (si es necesario).
*   **Generación de la energía sísmica:** Se activa la fuente sísmica (ej., golpe de martillo sobre la placa). El instante del impacto es el "tiempo cero" y es registrado por el sismógrafo.
*   **Registro de las ondas:** El sismógrafo registra las vibraciones captadas por cada geófono a lo largo del tiempo, generando un conjunto de trazas sísmicas (sismogramas).
*   **Apilamiento (Stacking):** Para mejorar la calidad de la señal y atenuar el ruido aleatorio, se repite la generación del disparo varias veces en el mismo punto, promediando los sismogramas obtenidos. Esto incrementa la relación señal/ruido ($\sqrt{\text{número de stacks}}$).
*   **Movimiento del punto de disparo:** Para un levantamiento de refracción, se realizan disparos en los extremos del arreglo y en puntos intermedios. Para MASW, la fuente se coloca a una distancia fija (*offset*) del primer geófono.
*   **Control de calidad en campo:** Durante la adquisición, es fundamental monitorear los sismogramas en tiempo real. Se debe verificar la claridad de las primeras llegadas (ondas P y S), la presencia de ruido excesivo o geófonos defectuosos. Si la calidad es pobre, se ajustan los parámetros de adquisición o se corrigen los problemas de acoplamiento.

### 4. Registro y Documentación

Una documentación exhaustiva es tan importante como la adquisición de los datos:

*   **Libreta de campo:** Registro de la fecha, hora, ubicación exacta (coordenadas GPS), tipo de equipo, configuración de los parámetros del sismógrafo, espaciado de geófonos, posición de los disparos, número de stacks, condiciones climáticas y del terreno, así como cualquier observación relevante o anomalía detectada (ruido, obstáculos, etc.).
*   **Fotos y esquemas:** Complementar con fotografías del sitio y esquemas de la disposición de los equipos.

### Consideraciones Finales

La fase de campo es exigente y requiere paciencia, atención al detalle y capacidad de adaptación a las condiciones cambiantes. Un equipo bien capacitado y la aplicación rigurosa de las mejores prácticas garantizan que los datos sísmicos adquiridos sean de la más alta calidad posible, sentando las bases para una interpretación geotécnica precisa y confiable.

## 2.3. Procesamiento e interpretación de resultados.

El éxito de cualquier investigación sísmica no solo radica en una meticulosa adquisición de datos en campo, sino también y de manera crucial, en un **procesamiento e interpretación de resultados** rigurosos. Esta fase transforma los datos crudos (sismogramas) en información geotécnica útil y modelos del subsuelo, lo cual requiere conocimientos especializados en geofísica y geotecnia.

### 1. Procesamiento de Datos Sísmicos

El procesamiento de datos sísmicos es una secuencia de operaciones que buscan mejorar la relación señal/ruido, corregir efectos no deseados y preparar los datos para la interpretación. Aunque los pasos específicos varían ligeramente según el método sísmico (refracción, MASW, reflexión), los principios generales son los siguientes:

*   **Control de Calidad Inicial:** Revisión de los sismogramas para identificar trazas ruidosas, geófonos defectuosos o disparos deficientes. Puede implicar la eliminación o edición de datos de baja calidad.
*   **Asignación de Primeras Llegadas (Picking):** Es uno de los pasos más críticos en refracción sísmica. Consiste en identificar y marcar manualmente o automáticamente el tiempo exacto en que la primera onda (usualmente la onda P) llega a cada geófono. Para MASW, se identifican los tiempos de llegada de las ondas superficiales.
*   **Apilamiento (Stacking):** Si se realizaron múltiples disparos en el mismo punto, se promedian para aumentar la coherencia de la señal y reducir el ruido aleatorio.
*   **Filtros:** Aplicación de filtros de frecuencia (banda de paso) para eliminar ruido de alta o baja frecuencia, y filtros de dominio espacial (ej., f-k) para atenuar ruido superficial o de aire.
*   **Correcciones Geométricas (corrección por *offset*):** Ajustes a los tiempos de llegada debido a la distancia entre la fuente y el geófono, para 'alinear' los datos.
*   **Análisis de Dispersión (para MASW):** En MASW, el procesamiento central implica construir la curva de dispersión de las ondas superficiales, que muestra cómo la velocidad de fase o grupo de las ondas varía con la frecuencia o la longitud de onda. Esta curva es la base para la inversión.
*   **Inversión Tomográfica o de Curvas de Dispersión:** Aplicación de algoritmos matemáticos complejos que transforman los tiempos de viaje (refracción) o las curvas de dispersión (MASW) en un modelo de velocidades de ondas (P o S) con la profundidad y/o lateralmente. Esto puede implicar métodos como la tomografía de refracción, el método de las capas recíprocas (Generalized Reciprocal Method, GRM) o algoritmos de inversión de la curva de dispersión.

### 2. Interpretación de Resultados

La interpretación es el proceso de traducir los modelos de velocidades de ondas (sísmicas P y S) o las secciones de resistividad en información geológica y geotécnica significativa. Aquí es donde la experiencia del especialista es fundamental:

*   **Correlación con Geología:** Los modelos de velocidades se comparan con la geología conocida del sitio (mapas geológicos, reportes existentes) para identificar formaciones rocosas, tipos de suelo y estructuras geológicas.
*   **Identificación de Interfaces Geotécnicas:** Los cambios abruptos en las velocidades sísmicas suelen indicar interfaces entre diferentes estratos, la profundidad de la roca madre, o la ubicación del nivel freático.
*   **Estimación de Parámetros Geotécnicos:**
    *   **Módulos Elásticos Dinámicos:** Las velocidades de ondas P ($V_P$) y S ($V_S$), junto con la densidad, permiten calcular módulos de Young dinámico ($E_d$), módulo de cizalla dinámico ($G_d$) y la relación de Poisson dinámica ($
u_d$). Estos parámetros son cruciales para el diseño de cimentaciones, análisis de vibraciones y evaluaciones de respuesta sísmica de sitios.
    *   **Clasificación Sísmica de Sitios:** Especialmente con $V_S$ de MASW, se determina el parámetro $V_{S30}$ (velocidad promedio de onda S en los primeros 30 metros), utilizado en códigos de construcción sismorresistente para clasificar la rigidez del sitio.
    *   **Compactación y Rigidez:** Valores altos de $V_P$ y $V_S$ pueden indicar suelos compactos o roca sana, mientras que valores bajos pueden señalar suelos blandos, sueltos o fracturados.
*   **Detección de Anomalías:** Se buscan zonas con velocidades anómalamente bajas (cavitaciones, suelos muy sueltos, zonas fracturadas) o altas (cuerpos rocosos intrusivos, suelos muy densos).
*   **Integración con Datos Directos:** Es fundamental integrar los resultados sísmicos con datos de perforaciones (sondeos), ensayos de penetración (SPT, CPT) y pruebas de laboratorio. Esto permite calibrar los modelos sísmicos, resolver ambigüedades y validar las interpretaciones, creando un modelo geotécnico más completo y fiable.

### 3. Consideraciones Clave en la Interpretación

*   **No Unicidad:** Es importante recordar que diferentes configuraciones geológicas pueden producir respuestas sísmicas similares. La interpretación siempre debe considerar múltiples hipótesis y usar toda la información disponible.
*   **Resolución:** La capacidad para detectar características depende de la resolución del método. Pequeñas anomalías o capas muy delgadas pueden no ser visibles, especialmente a gran profundidad.
*   **Contexto Geológico:** Una interpretación exitosa se apoya en un profundo conocimiento de la geología regional y local.
*   **Software Especializado:** Se utilizan paquetes de software específicos para el modelado directo e inverso de los datos sísmicos (ej., SeisImager, Geopsy, SURFER, etc.).

El procesamiento e interpretación de datos sísmicos es un arte y una ciencia que requiere no solo dominio técnico, sino también un juicio experto para transformar las ondas capturadas en información concreta y aplicable a los desafíos de la ingeniería geotécnica.

## 2.4. Diseño del informe de campo.

El **informe de campo** es un documento esencial que consolida toda la información recopilada durante la fase de adquisición de datos de un levantamiento geofísico. Su propósito es proporcionar una descripción clara y detallada de la metodología empleada, las condiciones del sitio, los datos brutos adquiridos y cualquier observación relevante que pueda influir en el procesamiento e interpretación posterior. Un informe de campo bien estructurado y exhaustivo es crítico para la trazabilidad y la reproducibilidad del estudio sísmico.

### Contenido Clave de un Informe de Campo Sísmico:

Un informe de campo debe ser lo suficientemente completo como para permitir que un tercero comprenda y, si es necesario, replique el trabajo realizado. Los elementos fundamentales incluyen:

1.  **Título y Datos del Proyecto:**
    *   Nombre del proyecto.
    *   Localización geográfica (país, región, coordenadas generales).
    *   Cliente.
    *   Fechas de la campaña de campo.
    *   Personal involucrado (jefe de equipo, operadores).

2.  **Objetivo del Estudio:**
    *   Breve descripción de la finalidad de la investigación sísmica (ej., caracterización geotécnica para cimentaciones, detección de cavidades, clasificación sísmica de sitios, etc.).

3.  **Descripción del Sitio:**
    *   **Localización detallada:** Croquis del área de estudio con puntos de referencia, coordenadas GPS precisas de los puntos de inicio y fin de cada línea sísmica.
    *   **Topografía:** Breve descripción de la morfología del terreno y presencia de desniveles significativos.
    *   **Geología superficial:** Descripción de la litología aflorante, presencia de suelos, vegetación, etc.
    *   **Condiciones ambientales:** Clima durante el levantamiento (temperatura, lluvia, viento), estado del suelo (seco, húmedo, inundado).
    *   **Condiciones de ruido:** Fuentes de ruido cultural (tráfico, maquinaria, líneas eléctricas, edificaciones cercanas) y su posible impacto en la calidad de los datos.

4.  **Metodología y Equipamiento:**
    *   **Método sísmico empleado:** Especificar si es refracción sísmica, MASW, reflexión, etc.
    *   **Equipamiento utilizado:**
        *   **Sismógrafo:** Marca, modelo y número de canales.
        *   **Geófonos:** Tipo (P, S o triaxiales), frecuencia natural (ej., 4.5 Hz), número de geófonos utilizados.
        *   **Fuente sísmica:** Tipo de fuente (martillo de 8 kg, *weight drop*, explosivo), placa de acoplamiento, sistema de *trigger* (martillo instrumentado, geófono de disparo).
        *   **Cables y accesorios:** Descripción general.
    *   **Configuración del levantamiento:**
        *   **Disposición del arreglo:** Configuración de los geófonos (lineal, 2D, etc.).
        *   **Espaciado de geófonos:** Distancia entre geófonos (ej., 2 m).
        *   **Longitud del arreglo:** Longitud total de cada línea (ej., 46 m para 24 geófonos a 2 m).
        *   **Posiciones de disparo (shots):** Ubicación de las fuentes sísmicas en relación con el arreglo de geófonos (ej., *end-on*, *mid-spread*, *off-set* para MASW). Número de disparos por línea.
        *   **Parámetros de adquisición:** Frecuencia de muestreo (ej., 0.25 ms), duración del registro (ej., 250 ms), número de *stacks* o apilamientos por disparo (ej., 5).

5.  **Registro de Datos en Campo:**
    *   **Inventario de datos brutos:** Listado de los archivos de datos generados, incluyendo nombres de archivo y fecha/hora de creación.
    *   **Observaciones relevantes por línea/disparo:** Cualquier incidencia durante la adquisición (ej., mal acoplamiento de geófonos, anomalías en los sismogramas, ruidos inesperados, interrupciones).
    *   **Esquemas o fotografías:** Imágenes que ilustren la disposición del equipo en cada línea sísmica y las condiciones del terreno.

6.  **Control de Calidad en Campo:**
    *   Descripción de las verificaciones realizadas durante la adquisición (ej., pruebas de continuidad, monitoreo visual de sismogramas, ajustes de ganancia).
    *   Identificación de líneas o disparos que puedan requerir un procesamiento especial o que se consideren de menor calidad.

7.  **Conclusiones de Campo (Preliminares):**
    *   Breve resumen de la calidad general de los datos y cualquier observación inicial sobre la geología o las anomalías detectadas en campo, sin entrar en detalles de interpretación profunda.

### Importancia del Informe de Campo:

El informe de campo es la base sobre la cual se construye todo el análisis geofísico. Una documentación deficiente puede llevar a interpretaciones erróneas, retrasos en el proyecto y, en última instancia, a decisiones de ingeniería comprometidas. Sirve como un registro permanente de las condiciones bajo las cuales se obtuvieron los datos y es una referencia invaluable para el equipo de procesamiento e interpretación, así como para futuros estudios o auditorías.

# 3. Aplicación de métodos eléctricos en geotecnia

## 3.1. Disposición, rendimiento y limitaciones del método eléctrico.

El **método eléctrico**, en particular la resistividad eléctrica, es una técnica geofísica ampliamente utilizada en geotecnia por su capacidad para detectar contrastes en las propiedades eléctricas del subsuelo, lo cual se correlaciona con la litología, el contenido de humedad, la porosidad, la salinidad y la presencia de arcillas. Comprender su disposición, rendimiento y limitaciones es fundamental para su aplicación efectiva.

### Disposición del Método Eléctrico (Sondeos Eléctricos Verticales - SEV y Tomografía Eléctrica - ERT)

La disposición se refiere a la geometría de los electrodos en el campo. La elección de la configuración de electrodos influye en la profundidad de investigación, la resolución lateral y la sensibilidad a diferentes estructuras.

1.  **Sondeos Eléctricos Verticales (SEV):**
    *   **Principio:** Mide la resistividad aparente del subsuelo variando la separación de los electrodos a partir de un punto central. Se busca un perfil de resistividad en profundidad.
    *   **Configuraciones comunes:**
        *   **Wenner:** Cuatro electrodos equiespaciados (A, M, N, B), donde la distancia entre A-M, M-N y N-B es la misma ('a'). Ofrece buena sensibilidad a las capas horizontales y es menos sensible al ruido lateral. La profundidad de investigación es aproximadamente igual a la distancia 'a'.
        *   **Schlumberger:** Los electrodos de corriente (A, B) se separan progresivamente, mientras que los electrodos de potencial (M, N) permanecen fijos o se mueven en pasos más pequeños en el centro. Permite investigar mayores profundidades con menos reubicaciones de electrodos de potencial, pero es más sensible al ruido.
    *   **Objetivo:** Determinar la profundidad y espesor de capas horizontales o subhorizontales, identificar el nivel freático, y caracterizar la resistividad de los materiales del subsuelo.
    *   **Disposición:** Se extiende una línea con los electrodos, manteniendo el centro fijo y expandiendo los electrodos exteriores para aumentar la profundidad de investigación.

2.  **Tomografía de Resistividad Eléctrica (ERT o REI):**
    *   **Principio:** Utiliza un gran número de electrodos (a menudo 24, 48 o más) conectados a un equipo multielectrodo. El sistema controla automáticamente la secuencia de inyección de corriente y medición de potencial, permitiendo obtener un gran número de mediciones con diferentes separaciones y configuraciones, construyendo un perfil 2D (o 3D) de resistividad.
    *   **Configuraciones comunes (para ERT):**
        *   **Wenner:** Similar al SEV, pero se realiza de forma multielectrodo a lo largo de una línea. Ofrece buena resolución vertical y lateral para estructuras compactas.
        *   **Dipolo-Dipolo:** Los dipolos de corriente y potencial se separan. Proporciona buena resolución lateral y es sensible a estructuras verticales, pero con menor relación señal/ruido a gran profundidad.
        *   **Polos-Dipolo/Polo-Polo:** Utilizan un electrodo de corriente o potencial 'infinito' alejado del arreglo. Ofrecen una buena penetración y sensibilidad a estructuras profundas, pero la logística de instalar un electrodo 'infinito' puede ser complicada.
        *   **Gradient:** Configuración mixta que permite investigar amplios volúmenes.
    *   **Objetivo:** Obtener imágenes 2D o 3D de la distribución de resistividades, ideal para la detección de cavidades, zonas de fractura, plumas de contaminación, intrusiones, nivel freático variable, y cartografía de formaciones geológicas complejas.
    *   **Disposición:** Los electrodos se instalan en línea recta (para 2D) o en una cuadrícula (para 3D) con espaciado constante, y el equipo conmuta automáticamente las mediciones.

### Rendimiento del Método Eléctrico

El rendimiento se evalúa en términos de resolución, precisión y aplicabilidad a diferentes problemas geotécnicos.

*   **Resolución:** La resolución de los métodos eléctricos es buena para detectar contrastes significativos en la resistividad. La ERT ofrece una resolución lateral y vertical superior a los SEV, permitiendo identificar geometrías complejas.
*   **Precisión:** La precisión de las mediciones depende de la calidad del acoplamiento de los electrodos, el ruido ambiental y la correcta calibración del equipo.
*   **Profundidad de Investigación:** Los SEV pueden alcanzar profundidades significativas (decenas a cientos de metros) expandiendo los electrodos. La ERT, con arreglos de 48 a 96 electrodos y separaciones de 5-10 m, puede investigar profundidades de hasta 50-100 m, dependiendo de la configuración y las condiciones del subsuelo.
*   **Información Geotécnica:** Muy sensible a la presencia de agua (especialmente salina), arcillas (baja resistividad), rocas fracturadas o meteorizadas (baja resistividad), roca sana (alta resistividad), y suelos arenosos o gravosos secos (alta resistividad). Permite mapear el nivel freático, detectar zonas de alteración, cavidades, e incluso estimar la porosidad y la permeabilidad de forma indirecta.

### Limitaciones del Método Eléctrico

Las limitaciones inherentes al método eléctrico deben ser consideradas cuidadosamente para evitar interpretaciones erróneas:

1.  **Principio de Equivalencia:** Diferentes combinaciones de espesores y resistividades de capas pueden producir la misma curva de resistividad aparente. Esto conduce a la no unicidad de la solución, requiriendo validación con datos directos.
2.  **Principio de Supresión:** Una capa de resistividad muy baja o muy alta puede 'enmascarar' la presencia de capas adyacentes, haciendo difícil su detección o caracterización precisa.
3.  **Sensibilidad a Fuentes de Ruido:** Las mediciones eléctricas son altamente sensibles a interferencias electromagnéticas generadas por líneas eléctricas, tuberías metálicas, cercas, y otros elementos conductivos cercanos, lo que puede degradar seriamente la calidad de los datos.
4.  **Necesidad de Buen Contacto de Electrodos:** Los electrodos deben tener un buen contacto eléctrico con el suelo para una medición precisa. Suelos secos, rocosos o con mucha vegetación pueden dificultar la instalación y el acoplamiento.
5.  **Profundidad Limitada en Suelos Muy Conductivos:** En terrenos con resistividades muy bajas (ej., arcillas marinas, suelos salinos), la señal se atenúa rápidamente, limitando la profundidad de investigación efectiva.
6.  **Interpretación Subjetiva:** La conversión de resistividades en propiedades geotécnicas es indirecta y requiere experiencia y correlación con datos de perforación y muestreo.
7.  **Condiciones del Terreno:** La topografía abrupta puede introducir errores en las mediciones y dificultar la modelización 2D/3D. La presencia de cuerpos metálicos enterrados también puede generar anomalías espurias.
8.  **Costos de Campo (ERT):** Aunque más eficiente que los SEV individuales, la instalación de un gran número de electrodos para ERT puede ser laboriosa y tomar tiempo, especialmente en terrenos difíciles.

En resumen, los métodos eléctricos son herramientas valiosas para la caracterización geotécnica del subsuelo, especialmente en la detección de agua y variaciones litológicas. Sin embargo, su aplicación exitosa depende de una cuidadosa planificación de la disposición, una adecuada gestión del ruido, la consideración de sus limitaciones y, fundamentalmente, la calibración y validación con datos geotécnicos directos.

## 3.2. Instalación de equipos y toma de datos en el campo.

La **instalación de equipos y la toma de datos en el campo** para los métodos eléctricos, en particular para los Sondeos Eléctricos Verticales (SEV) y la Tomografía de Resistividad Eléctrica (ERT), son procedimientos que, si bien comparten principios con otras técnicas geofísicas, tienen sus particularidades. La calidad del contacto entre los electrodos y el terreno es crucial para obtener mediciones fiables.

### 1. Planificación Previa a la Campaña de Campo

Como en cualquier levantamiento geofísico, una planificación detallada es indispensable:

*   **Reconocimiento del sitio:** Evaluar la accesibilidad, topografía, geología superficial, presencia de posibles fuentes de ruido electromagnético (líneas eléctricas, tuberías metálicas, cercas, estructuras), y la humedad del suelo.
*   **Diseño del levantamiento:** Definir la ubicación y longitud de las líneas de SEV o ERT, el espaciado entre electrodos, y las configuraciones de arreglo más adecuadas en función de los objetivos (profundidad de investigación, resolución esperada).
*   **Selección de equipos:** Elegir el resistivímetro (conmutador multielectrodo para ERT), los electrodos (varillas metálicas de acero inoxidable o cobre), cables, y una fuente de energía (batería, generador) si el equipo lo requiere. Asegurarse de tener piquetas y martillos para la instalación de los electrodos.
*   **Permisos y seguridad:** Obtener los permisos necesarios y establecer protocolos de seguridad.

### 2. Instalación del Equipo en Campo

La fase de instalación requiere atención al detalle para asegurar un buen acoplamiento eléctrico:

*   **Establecimiento de las líneas:** Marcar la ubicación de cada electrodo con precisión a lo largo de la línea definida por el diseño. Utilizar un GPS para registrar las coordenadas de inicio y fin, y puntos intermedios si la línea es larga o irregular.
*   **Instalación de electrodos:** Las varillas metálicas se clavan firmemente en el suelo a la profundidad necesaria para asegurar un buen contacto eléctrico. Es fundamental que la resistencia de contacto sea lo más baja posible. En suelos secos o muy rocosos, puede ser necesario humedecer el suelo alrededor del electrodo con agua salada o utilizar bentonita para mejorar el contacto.
*   **Conexión de cables:** Los cables aislados se extienden a lo largo de la línea y se conectan a cada electrodo y al resistivímetro. Se debe evitar la creación de bucles de cable que puedan actuar como antenas y captar ruido electromagnético. Se verifican las conexiones para asegurar la continuidad y ausencia de cortocircuitos.
*   **Conexión al resistivímetro:** El equipo se ubica en un lugar central y seguro, alejado de fuentes de ruido si es posible.

### 3. Toma de Datos (Adquisición Eléctrica)

El proceso de adquisición varía según el método:

*   **Para Sondeos Eléctricos Verticales (SEV):**
    *   **Configuración manual o semiautomática:** Se establece la configuración de electrodos (Wenner, Schlumberger) y se inyecta corriente entre los electrodos de corriente (A, B) mientras se mide la diferencia de potencial entre los electrodos de potencial (M, N).
    *   **Expansión del arreglo:** Se van aumentando progresivamente las separaciones entre los electrodos de corriente (A, B) y potencial (M, N) de forma simétrica con respecto al centro del sondeo. Esto permite investigar mayores profundidades.
    *   **Registro de datos:** Para cada separación, se registra la corriente inyectada, la diferencia de potencial medida y el factor geométrico K (que depende de la configuración y la distancia entre electrodos) para calcular la resistividad aparente.
*   **Para Tomografía de Resistividad Eléctrica (ERT):**
    *   **Configuración automatizada:** Con un resistivímetro multielectrodo, se conectan todos los electrodos a un cable común, y el equipo conmuta automáticamente las conexiones para inyectar corriente y medir potencial según una secuencia programada (ej., Wenner, Dipolo-Dipolo, etc.).
    *   **Barrido automático:** El equipo realiza un barrido completo de todas las combinaciones de mediciones posibles o predefinidas para la configuración y el número de electrodos dados, almacenando la resistividad aparente y la información de los electrodos de corriente y potencial.
    *   **Control de calidad en campo:** Durante la adquisición, el equipo suele mostrar la resistencia de contacto de los electrodos y el error de ajuste. Si la resistencia de contacto es muy alta o el error es significativo, se debe detener la medición y mejorar el contacto del electrodo problemático (ej., re-clavar, humedecer el suelo). También es importante monitorear la forma de onda de la señal y la relación señal/ruido para evitar datos contaminados.

### 4. Registro y Documentación

Una documentación detallada del trabajo de campo es esencial:

*   **Libreta de campo:** Anotar la fecha, hora, ubicación exacta de las líneas/sondeos (coordenadas GPS), tipo de equipo, configuración de los parámetros del resistivímetro, espaciado de electrodos, condiciones del terreno (humedad, vegetación), clima, y cualquier observación o incidencia (ej., electrodos con mal contacto, fuentes de ruido, obstáculos).
*   **Diagramas y fotografías:** Esquemáticos de la disposición de los electrodos y el equipo, así como fotografías del sitio para contextualizar los datos.

### Consideraciones Adicionales

*   **Acoplamiento de electrodos:** La baja resistencia de contacto es fundamental. En suelos muy secos o helados, la adquisición puede ser muy difícil o imposible.
*   **Interferencias:** Mantener los cables alejados de líneas eléctricas o estructuras metálicas para minimizar el ruido.
*   **Profundidad vs. Resolución:** La separación de los electrodos controla tanto la profundidad de investigación como la resolución. Arreglos más amplios permiten mayor penetración pero con menor detalle, mientras que arreglos más densos ofrecen mayor resolución a menor profundidad.

Una ejecución precisa de la instalación y adquisición de datos asegura que la información eléctrica recolectada sea de la más alta calidad, sentando las bases para un procesamiento e interpretación posterior exitosos en el contexto de la ingeniería geotécnica.

## 3.3. Procesamiento, interpretación de resultados, y correlación de métodos.

Una vez que los datos de resistividad eléctrica han sido adquiridos en campo con la máxima calidad, la siguiente etapa crucial es el **procesamiento, la interpretación de los resultados y, fundamentalmente, la correlación de estos métodos** con otras fuentes de información geotécnica. Este proceso transforma las mediciones de campo en un modelo comprensible del subsuelo, esencial para la toma de decisiones en proyectos de ingeniería civil y geotécnica.

### 1. Procesamiento de Datos Eléctricos

El procesamiento de los datos eléctricos, especialmente para la Tomografía de Resistividad Eléctrica (ERT), implica una serie de pasos para eliminar ruido y generar un modelo geológico-geotécnico del subsuelo:

*   **Control de Calidad:** Revisión inicial de los datos brutos. Se eliminan las mediciones de baja calidad, aquellas con altos errores de contacto de electrodos o aquellas afectadas por ruido externo significativo. Los equipos modernos suelen realizar este filtrado automáticamente durante la adquisición o permiten identificar y eliminar puntos anómalos.
*   **Inversión:** Este es el paso central. La inversión es un proceso numérico iterativo que transforma las mediciones de resistividad aparente de campo en un modelo 2D (o 3D) de resistividad real del subsuelo. Se utilizan algoritmos que minimizan la diferencia entre las resistividades aparentes medidas y las calculadas a partir de un modelo hipotético, ajustando este modelo hasta que la diferencia sea aceptable. Los programas de inversión populares incluyen RES2DINV y EarthImager 2D/3D.
    *   **Objetivo de la inversión:** Obtener un modelo de distribución de resistividades reales que sea consistente con las mediciones de campo y, a la vez, geológicamente plausible. El resultado es una "imagen" del subsuelo que muestra cómo varía la resistividad con la profundidad y lateralmente.

### 2. Interpretación de Resultados

La interpretación es el paso en el que el modelo de resistividad se traduce en información geológica y geotécnica relevante. Esta etapa requiere un conocimiento profundo de la geofísica, la geología y la geotecnia del área de estudio:

*   **Correlación Litológica:** Las resistividades de los materiales del subsuelo varían ampliamente. Por ejemplo:
    *   **Bajas resistividades:** Suelen indicar presencia de arcillas, suelos saturados con agua salina, zonas muy fracturadas o meteorizadas, o presencia de contaminantes iónicos.
    *   **Altas resistividades:** Indican rocas sanas y compactas, gravas y arenas secas, suelos arenosos, o cuerpos de aire (cavidades secas).
    El modelo de resistividad se compara con la experiencia local, la geología regional y la bibliografía para asignar probables litologías a las diferentes zonas de resistividad.
*   **Identificación de Estructuras Geotécnicas:**
    *   **Nivel Freático:** A menudo se observa un cambio abrupto en la resistividad al pasar de la zona no saturada (mayor resistividad) a la zona saturada (menor resistividad, dependiendo del tipo de suelo y la salinidad del agua).
    *   **Zonas de Alteración/Fracturación:** Las zonas con mayor presencia de agua o arcillas dentro de un macizo rocoso (baja resistividad) pueden indicar fracturación o intensa meteorización.
    *   **Cavidades/Vacíos:** Pueden manifestarse como anomalías de muy alta resistividad (si están secas) o muy baja (si están llenas de agua conductiva).
    *   **Diques/Intrusiones:** Pueden mostrar contrastes de resistividad si su composición es diferente al material encajante.
    *   **Plumas de Contaminación:** Ciertas plumas (ej., salinas) pueden alterar la conductividad eléctrica del subsuelo, creando anomalías detectables.
*   **Delimitación de Unidades:** El modelo permite delimitar la extensión vertical y lateral de diferentes unidades geotécnicas.

### 3. Correlación de Métodos

La fortaleza de la geofísica aplicada en geotecnia a menudo reside en la **correlación y la integración de múltiples métodos**. Un solo método geofísico raramente proporciona una solución única y completa debido a la inherente no unicidad de las soluciones inversas y a la dependencia de diferentes propiedades físicas. La combinación con otros métodos geofísicos y, crucialmente, con datos directos es esencial:

*   **Correlación con Sondeos y Calicatas:** Los puntos donde se realizaron perforaciones o calicatas son fundamentales. Permiten:
    *   **Calibración:** Asignar rangos de resistividad a litologías específicas o a la presencia de agua/arcillas obtenidas directamente del sondeo.
    *   **Validación:** Confirmar la profundidad de interfaces o la presencia de anomalías detectadas por el método eléctrico.
    *   **Reducción de Ambigüedad:** Resolver interpretaciones múltiples, dado que los datos directos ofrecen una "verdad en el terreno".
*   **Correlación con Otros Métodos Geofísicos:**
    *   **Sísmica (Refracción/MASW):** La combinación de resistividad y velocidades sísmicas es muy potente. Por ejemplo, una zona de baja resistividad y baja velocidad sísmica podría indicar un suelo blando o muy fracturado y saturado, mientras que una baja resistividad con alta velocidad sísmica podría sugerir una arcilla muy plástica pero competente o una roca alterada con arcilla en las fracturas. Una zona de alta resistividad y alta velocidad sísmica apuntaría a una roca sana.
    *   **Electromagnéticos (EM):** Métodos como el EM-31 o EM-34 miden la conductividad, que es el inverso de la resistividad. Pueden ser complementarios para cubrir grandes áreas rápidamente.
    *   **GPR (Ground Penetrating Radar):** Aunque mide la permitividad dieléctrica, a menudo se correlaciona con la resistividad, siendo muy efectivo para alta resolución en el subsuelo superficial, mientras que los métodos de resistividad penetran más profundamente.

La correlación permite construir un **modelo geotécnico integrado y más robusto**, donde las fortalezas de un método compensan las debilidades de otro, reduciendo la incertidumbre y proporcionando una base sólida para la ingeniería geotécnica. Un informe final debe presentar no solo los resultados geofísicos, sino también cómo estos se integran con el conocimiento geológico y geotécnico del sitio.

## 3.4. Contenido de informes de campo.

El **informe de campo** para levantamientos eléctricos es un documento crucial que resume de manera organizada toda la información recopilada durante la fase de adquisición de datos en campo. Su objetivo principal es asegurar la trazabilidad del trabajo, documentar las condiciones de adquisición y proporcionar la base para el procesamiento e interpretación posterior. Un informe de campo completo y preciso es indispensable para garantizar la calidad y fiabilidad de los resultados finales.

### Elementos Fundamentales del Informe de Campo para Métodos Eléctricos:

Un informe de campo debe incluir los siguientes componentes para ser considerado exhaustivo:

1.  **Datos Generales del Proyecto:**
    *   **Identificación del Proyecto:** Nombre, código, o referencia del proyecto.
    *   **Cliente:** Nombre de la entidad o persona que encarga el estudio.
    *   **Ubicación del Sitio:** Dirección, municipio, provincia, país. Puede incluir un mapa de localización general.
    *   **Fechas de la Campaña:** Período exacto de realización de los trabajos de campo.
    *   **Personal de Campo:** Nombres y roles de los integrantes del equipo que participó en la adquisición de datos.

2.  **Objetivos Específicos del Levantamiento Eléctrico:**
    *   Breve descripción de lo que se busca caracterizar o detectar con el método eléctrico (ej., delimitación de nivel freático, detección de cavidades, cartografía de plumas de contaminación, identificación de litologías, estudio de cimentaciones, etc.).

3.  **Descripción Detallada del Sitio:**
    *   **Topografía:** Características del relieve (plano, inclinado, accidentado). Se pueden incluir perfiles topográficos o curvas de nivel si se tienen.
    *   **Geología Superficial:** Descripción de los materiales expuestos, tipo de suelo, presencia de roca, vegetación. Se puede referenciar mapas geológicos regionales.
    *   **Hidrografía:** Cuerpos de agua cercanos, drenajes, humedad general del suelo.
    *   **Infraestructuras y Obstáculos:** Identificación de caminos, edificaciones, líneas eléctricas, tuberías (metálicas o no), cercas, y cualquier elemento que pueda interferir con las mediciones o limitar la disposición de los electrodos.
    *   **Fuentes de Ruido:** Descripción de posibles fuentes de ruido electromagnético (líneas de transmisión de energía, transformadores, industrias, vías férreas) y su proximidad a las líneas de medición.
    *   **Condiciones Climáticas:** Clima durante la adquisición (temperatura, precipitaciones, humedad ambiental), ya que pueden afectar la resistividad del suelo y la resistencia de contacto de los electrodos.

4.  **Metodología y Equipamiento Utilizado:**
    *   **Método Eléctrico:** Especificar si se realizaron Sondeos Eléctricos Verticales (SEV), Tomografía de Resistividad Eléctrica (ERT 2D/3D), u otro método relacionado (IP, SP).
    *   **Equipamiento Principal:**
        *   **Resistivímetro/Georesistivímetro:** Marca, modelo, número de canales (para ERT).
        *   **Electrodos:** Tipo de material (acero inoxidable, cobre), diámetro, longitud. Número total de electrodos utilizados para ERT.
        *   **Cables:** Tipo, longitud, conectores.
        *   **GPS:** Marca y modelo del sistema de posicionamiento utilizado.
        *   **Otros accesorios:** Martillos, piquetas, baldes (para humedecer).
    *   **Configuración del Levantamiento por Línea/Punto:**
        *   **Tipo de Arreglo:** Wenner, Schlumberger, Dipolo-Dipolo, etc.
        *   **Espaciado de Electrodos:** Distancia constante entre electrodos para ERT (ej., 5 m). Para SEV, las configuraciones de expansión.
        *   **Longitud de las Líneas/Arreglos:** Extensión total de cada perfil ERT o SEV.
        *   **Orientación:** Dirección de las líneas (ej., N-S, E-W) y coordenadas de inicio y fin de cada línea (UTM, Lat/Lon).
        *   **Parámetros de Adquisición:** Número de ciclos de medida, tiempo de integración, corriente máxima inyectada, control de calidad del equipo (ej., error de RMS para ERT).

5.  **Registro de Datos de Campo:**
    *   **Tabla de Líneas/SEV:** Un listado de todas las líneas de ERT o puntos de SEV realizados, con su identificación, coordenadas, longitud, espaciado de electrodos y número de archivo de datos brutos generado.
    *   **Inventario de Archivos:** Nombres y formatos de los archivos de datos brutos guardados por el equipo.
    *   **Observaciones Críticas:** Detalles sobre cualquier problema durante la adquisición (ej., electrodos con alta resistencia de contacto, necesidad de humedecer el suelo, presencia de ruido inesperado que pudo afectar ciertas mediciones, fallas del equipo, interrupciones).

6.  **Esquemas y Fotografía:**
    *   **Croquis del Levantamiento:** Un esquema general que muestre la disposición de todas las líneas de medición en el sitio.
    *   **Fotos del Sitio:** Imágenes que capturen las condiciones generales del terreno, la instalación típica de los electrodos, la presencia de obstáculos o fuentes de ruido, y el equipo en operación.

7.  **Notas y Comentarios Adicionales:**
    *   Cualquier otra información que se considere relevante para el procesamiento e interpretación, como cambios en la metodología debido a condiciones imprevistas, o sugerencias para el equipo de procesamiento.

El informe de campo no solo valida el trabajo realizado, sino que también actúa como una guía esencial para los especialistas que posteriormente procesarán e interpretarán los datos, permitiéndoles comprender el contexto de cada medición y tomar decisiones informadas sobre la calidad y el manejo de la información obtenida.

# 4. Gravimetría y magnetometría en geotecnia

## 4.1. Principios de los métodos gravimétricos y magnéticos.

Los métodos geofísicos gravimétricos y magnéticos, a menudo agrupados por su naturaleza pasiva, son herramientas valiosas en la exploración del subsuelo, complementando otras técnicas activas. Ambos se basan en la medición de variaciones de campos físicos naturales (gravitatorio y magnético, respectivamente) que son perturbados por la presencia de cuerpos o estructuras con propiedades físicas contrastantes en el subsuelo. Su aplicación en geotecnia e ingeniería civil permite la identificación de anomalías que pueden ser indicativas de características geotécnicas relevantes.

### 1. Principios de los Métodos Gravimétricos

El **método gravimétrico** se fundamenta en la **Ley de Gravitación Universal de Isaac Newton**, que establece que dos cuerpos cualesquiera se atraen con una fuerza directamente proporcional al producto de sus masas e inversamente proporcional al cuadrado de la distancia que los separa. En el contexto geofísico, esta ley se aplica para estudiar las variaciones espaciales en la **aceleración de la gravedad (g)** sobre la superficie terrestre.

*   **Base Física:** La aceleración de la gravedad en un punto dado de la superficie terrestre no es uniforme. Está influenciada principalmente por:
    *   **La masa total de la Tierra y su rotación:** Que dan lugar a la gravedad teórica o normal.
    *   **La distribución de las masas en el subsuelo:** Variaciones en la densidad de los materiales rocosos y sedimentarios generan pequeñas perturbaciones o **anomalías gravimétricas** en el campo de gravedad normal. Una masa más densa genera una atracción gravitatoria localmente mayor (anomalía positiva), mientras que una masa menos densa (ej., una cavidad, un suelo suelto, un cuerpo de agua subterránea) genera una atracción menor (anomalía negativa).

*   **Propiedad Física Principal:** La **densidad** de los materiales del subsuelo. Los contrastes de densidad entre diferentes unidades geológicas o entre una anomalía y el material circundante son los que permiten la detección.

*   **Medición:** Se utilizan **gravímetros** de alta precisión para medir la aceleración de la gravedad en diferentes puntos de la superficie. Estos instrumentos son extremadamente sensibles y miden variaciones del orden de microgalios (1 Gal = 1 cm/s²).

*   **Aplicaciones en Geotecnia:**
    *   **Detección de Cavidades y Vacíos:** Zonas de muy baja densidad (cuevas, túneles abandonados, rellenos sueltos) generan anomalías gravimétricas negativas.
    *   **Mapeo de Contactos Geológicos:** Permite identificar la interfaz entre materiales con diferentes densidades, como roca fresca y roca alterada, o entre diferentes formaciones litológicas.
    *   **Estimación de Espesores de Depósitos Blandos:** La variación de densidad entre sedimentos y el basamento rocoso puede utilizarse para determinar la geometría del subsuelo.

*   **Limitaciones:** Las anomalías gravimétricas son

inherentemente ambiguas (múltiples distribuciones de masa pueden producir la misma anomalía). Requieren correcciones complejas (marea, deriva instrumental, latitud, topografía, Bouguer) para aislar la señal de interés geotécnico.

### 2. Principios de los Métodos Magnéticos (Magnetometría)

El **método magnético**, o **magnetometría**, se basa en la medición de las variaciones locales del **campo magnético terrestre** que son causadas por la presencia de materiales con diferente **susceptibilidad magnética** en el subsuelo.

*   **Base Física:** La Tierra posee un campo magnético principal generado por el movimiento del hierro fundido en su núcleo externo. Este campo principal es perturbado por:
    *   **Inducción Magnética:** La mayoría de los materiales del subsuelo adquieren una magnetización inducida al ser expuestos al campo magnético terrestre. La magnitud de esta magnetización depende de su **susceptibilidad magnética**; materiales con alta susceptibilidad (ej., minerales de hierro como la magnetita) se magnetizan más intensamente y, por lo tanto, distorsionan el campo localmente.
    *   **Magnetización Remanente:** Algunos materiales geológicos (especialmente rocas ígneas y metamórficas) conservan una magnetización adquirida en el pasado (magnetización remanente), que puede no estar alineada con el campo actual y contribuir a las anomalías.

*   **Propiedad Física Principal:** La **susceptibilidad magnética** de los materiales del subsuelo. Un alto contraste de susceptibilidad entre una anomalía y el entorno es necesario para la detección.

*   **Medición:** Se emplean **magnetómetros** (ej., magnetómetros de precesión de protones, de vapor atómico) para medir la intensidad total del campo magnético (o sus componentes) en diferentes puntos. Estos instrumentos son capaces de detectar variaciones muy pequeñas, del orden de nanoTeslas (nT).

*   **Aplicaciones en Geotecnia:**
    *   **Detección de Objetos Metálicos Enterrados:** Tuberías, tanques, residuos metálicos, cables, etc., debido a su alta susceptibilidad magnética, generan fuertes anomalías.
    *   **Mapeo de Fallas y Contactos Geológicos:** Si las estructuras geológicas o los contactos litológicos implican materiales con diferente contenido de minerales magnéticos (ej., una falla que separa dos formaciones con distinta magnetita).
    *   **Localización de Ciertas Rocas:** Cartografía de intrusiones ígneas o cuerpos metamórficos con minerales ferromagnéticos.
    *   **Detección de Zonas Contaminadas:** En algunos casos, la contaminación puede alterar el estado de oxidación de los minerales del suelo, afectando su respuesta magnética.

*   **Limitaciones:** La interpretación puede ser compleja debido a la superposición de anomalías (debido a la geometría dipolar de las fuentes magnéticas) y la ambigüedad en la fuente de la anomalía. Es muy susceptible a ruido cultural (líneas eléctricas, vehículos, cercas metálicas) y variaciones diarias del campo magnético terrestre (requiriendo estaciones base para corrección).

Ambos métodos son fundamentales para la caracterización del subsuelo, especialmente en fases iniciales de proyectos geotécnicos, proporcionando información valiosa sobre la distribución de masas y materiales con contrastes de propiedades, que luego puede ser complementada y validada con otros métodos geofísicos y directos.

## 4.2. Equipos de mediciones.

La eficacia de los métodos gravimétricos y magnéticos en geotecnia depende en gran medida de la precisión y sensibilidad de los equipos de medición utilizados. Estos instrumentos están diseñados para detectar variaciones extremadamente sutiles en los campos gravitatorio y magnético terrestre, que son indicativas de anomalías en el subsuelo. La elección y el correcto manejo del equipo son cruciales para obtener datos fiables.

### 1. Equipos para Métodos Gravimétricos (Gravímetros)

Los instrumentos utilizados para medir la aceleración de la gravedad se denominan **gravímetros**. Su diseño es complejo, ya que deben ser capaces de detectar variaciones de la gravedad de solo unas pocas partes por billón.

*   **Gravímetros Relativos:** Son los más comunes en exploración geofísica y geotécnica. Miden las diferencias relativas en la gravedad entre dos puntos. Su principio se basa en una masa suspendida de un resorte. Un cambio en la gravedad provoca un cambio en la longitud del resorte o en el torque ejercido sobre la masa, lo cual es detectado y cuantificado.
    *   **Tipos principales:**
        *   **Gravímetros de resorte metálico (ej. LaCoste & Romberg):** Utilizan un sistema de resorte de cuarzo o metal para suspender una masa. Son extremadamente sensibles y estables, pero susceptibles a la deriva (cambio lento y continuo en la lectura con el tiempo) que debe ser corregida mediante la repetición de mediciones en una estación base.
        *   **Gravímetros de resorte de cuarzo (ej. Worden):** Similares a los de resorte metálico, pero con un sistema de suspensión de cuarzo que les confiere mayor estabilidad térmica y menor deriva.
    *   **Características clave:**
        *   **Sensibilidad:** Del orden de microgalios (µGal), donde 1 Gal = 1 cm/s² = 1000 mGal = $10^6$ µGal. Las anomalías geotécnicas suelen ser de unas pocas decenas o cientos de µGal.
        *   **Estabilidad:** Capacidad de mantener una lectura constante en ausencia de cambios en la gravedad. La deriva es un factor importante a considerar.
        *   **Portabilidad:** Los gravímetros de campo son diseñados para ser transportables y robustos.

*   **Gravímetros Absolutos:** Miden el valor absoluto de la aceleración de la gravedad. Suelen basarse en la medición precisa del tiempo de caída libre de una masa en un vacío. Son instrumentos de laboratorio o de referencia, utilizados para calibrar gravímetros relativos o establecer estaciones de control en redes gravimétricas.

*   **Aplicaciones en Geotecnia:** Los gravímetros son utilizados en geotecnia para la detección de cavidades, túneles, o zonas de baja densidad (rellenos sueltos, zonas kársticas), donde se espera una anomalía de gravedad negativa. La alta sensibilidad es fundamental para detectar estas pequeñas perturbaciones.

### 2. Equipos para Métodos Magnéticos (Magnetómetros)

Los **magnetómetros** son instrumentos diseñados para medir la intensidad y, en algunos casos, la dirección del campo magnético terrestre y sus variaciones locales. Los magnetómetros utilizados en geofísica son de alta precisión, capaces de detectar anomalías que pueden ser de unas pocas unidades a miles de nanoTeslas (nT), en un campo terrestre total que ronda los 50.000 nT.

*   **Tipos principales:**
    *   **Magnetómetros de Precesión de Protones (PPM):** Son los más comunes para estudios geotécnicos. Miden la intensidad total del campo magnético (F). Su principio se basa en la precesión de los protones en un fluido rico en hidrógeno (agua, queroseno) al ser expuestos a un campo magnético. Un pulso de corriente alinea los momentos magnéticos de los protones; al cesar el pulso, los protones precesionan alrededor de la dirección del campo magnético terrestre con una frecuencia directamente proporcional a la intensidad de este campo.
        *   **Ventajas:** Alta precisión (±0.1 nT), estabilidad, bajo costo y fácil operación.
        *   **Limitaciones:** Solo miden la intensidad total del campo, no la dirección.

    *   **Magnetómetros de Vapor Atómico (Overhauser, Cesio):** Ofrecen mayor sensibilidad y velocidad de muestreo que los PPM, alcanzando precisiones de 0.01 nT o menos. Utilizan la absorción de luz por átomos de vapor (como cesio o rubidio) en presencia de un campo magnético. Son ideales para mediciones en movimiento (ej., en aeronaves o vehículos) y para detectar anomalías muy sutiles.
        *   **Ventajas:** Muy alta sensibilidad y velocidad, lo que permite detectar anomalías débiles y realizar mediciones continuas.
        *   **Limitaciones:** Mayor costo y complejidad en la operación.

    *   **Magnetómetros Fluxgate:** Miden las componentes del campo magnético (ej., componente vertical o horizontal). Se basan en la saturación magnética de un núcleo ferromagnético. Aunque han sido superados en precisión por otros tipos para mediciones de intensidad total, todavía se utilizan para mediciones de gradientes o para determinar la dirección del campo.

*   **Componentes Adicionales:**
    *   **Estación Base:** Para corregir las variaciones diarias del campo magnético terrestre, se utiliza un magnetómetro fijo (estación base) que registra continuamente el campo en un punto de referencia mientras se realizan las mediciones en campo. Esto permite eliminar el ruido temporal.
    *   **GPS Integrado:** Muchos magnetómetros modernos integran un receptor GPS para registrar automáticamente las coordenadas de cada medición, facilitando el mapeo de las anomalías.

*   **Aplicaciones en Geotecnia:** Los magnetómetros son ampliamente utilizados para la detección de objetos metálicos enterrados (tuberías, tambores, cables, escombros ferrosos), cartografía de fallas (si están mineralizadas con materiales magnéticos), o la localización de ciertos tipos de rocas (ej., intrusiones ígneas con alto contenido de magnetita).

En ambos casos, la capacitación del personal, la calibración regular de los equipos y la aplicación de protocolos de campo rigurosos son tan importantes como la calidad intrínseca de los instrumentos para garantizar resultados confiables en la exploración geotécnica.

## 4.3. Las correcciones de los métodos gravimétricos y magnéticos.

Los datos brutos obtenidos de las mediciones gravimétricas y magnéticas en campo no reflejan directamente las anomalías causadas por las heterogeneidades del subsuelo que interesan a la geotecnia. Estos datos están contaminados por una serie de efectos externos o de gran escala que deben ser eliminados mediante un proceso sistemático de **correcciones**. El objetivo de estas correcciones es aislar la señal de interés (la anomalía local) de los efectos no deseados, permitiendo una interpretación precisa.

### 1. Correcciones para Métodos Gravimétricos

Las mediciones de la aceleración de la gravedad (g) son influenciadas por múltiples factores además de la distribución de masas en el subsuelo local. Para obtener la **anomalía de Bouguer**, que es la que se interpreta en geotecnia, se deben aplicar las siguientes correcciones:

*   **Corrección por Deriva (Drift Correction):** Los gravímetros, especialmente los de resorte, tienden a sufrir una variación lenta y continua en su lectura con el tiempo debido a cambios de temperatura, deformaciones elásticas del resorte o fatiga del material. Esta corrección se aplica registrando el valor de la gravedad en una estación base al inicio y al final de la jornada de trabajo, y periódicamente durante el levantamiento. Asumiendo una deriva lineal o polinómica, se distribuye proporcionalmente a los tiempos de medición.

*   **Corrección por Marea Terrestre (Tidal Correction):** La atracción gravitatoria de la Luna y el Sol provoca deformaciones en la Tierra (mareas sólidas) que afectan el valor medido de la gravedad. Esta corrección se calcula con modelos teóricos basados en la posición de los cuerpos celestes y la latitud del punto de medición. Aunque pequeña (unas pocas decenas de µGal), es importante para trabajos de alta precisión.

*   **Corrección por Latitud (Latitude Correction):** La Tierra no es una esfera perfecta, sino un elipsoide achatado en los polos, y su rotación genera una fuerza centrífuga que varía con la latitud. Por lo tanto, la gravedad aumenta desde el ecuador hacia los polos. Esta corrección elimina la variación teórica de la gravedad en un elipsoide de referencia (ej., GRS80) y se calcula en función de la latitud de cada estación de medición.

*   **Corrección por Aire Libre (Free-Air Correction - FAC):** Esta corrección compensa el hecho de que las estaciones de medición no están todas a la misma elevación. La gravedad disminuye con la altura sobre el nivel del mar. La FAC suma un valor positivo a las estaciones elevadas y resta a las más bajas para referir todas las mediciones a un datum común (generalmente el geoide). El gradiente de aire libre es aproximadamente $-0.3086 \text{ mGal/m}$.

*   **Corrección de Bouguer (Bouguer Correction - BC):** Esta es una de las correcciones más importantes. Elimina la atracción gravitatoria de la masa de roca o suelo entre la estación de medición y el datum (generalmente el geoide). Se calcula asumiendo una lámina horizontal infinita de densidad uniforme. La fórmula es $2\pi G \rho h$, donde $G$ es la constante gravitacional, $\rho$ es la densidad de Bouguer (densidad promedio de los materiales superficiales) y $h$ es la altura de la estación sobre el datum. Un valor inadecuado de $\rho$ puede introducir anomalías artificiales.

*   **Corrección Topográfica o de Terreno (Terrain Correction - TC):** La corrección de Bouguer asume una lámina infinita, es decir, un terreno plano. Sin embargo, la topografía circundante (colinas cercanas o valles) ejerce una atracción o un efecto de 'alivio' que afecta la lectura del gravímetro. Esta corrección es siempre positiva (ya que tanto las masas elevadas como las deprimidas reducen la gravedad medida) y se calcula dividiendo el terreno circundante en zonas (zonas de Hayford) y aplicando un algoritmo basado en la topografía local y regional. Es particularmente crítica en zonas de relieve pronunciado.

Después de aplicar todas estas correcciones, se obtiene la **Anomalía de Bouguer**, que representa las variaciones de gravedad causadas por las heterogeneidades de densidad en el subsuelo por debajo del datum de referencia. Esta anomalía es la que se utiliza para la interpretación geotécnica.

### 2. Correcciones para Métodos Magnéticos

Las mediciones del campo magnético también requieren correcciones para aislar las anomalías de interés geotécnico de las variaciones de gran escala o temporales. El objetivo es obtener una **anomalía magnética residual**.

*   **Corrección por Variación Diurna (Diurnal Correction):** El campo magnético terrestre varía a lo largo del día debido a la interacción de los vientos solares con la ionosfera. Estas variaciones pueden ser del orden de decenas o incluso cientos de nT, y enmascarar las anomalías del subsuelo. Para corregir esto, se utiliza una **estación base** magnética que registra continuamente el campo magnético en un punto fijo cerca del área de estudio. Las lecturas de la estación base se restan de las mediciones de campo para eliminar esta variación temporal.

*   **Corrección por Campo Geomagnético Regional (Regional Field Correction):** El campo magnético terrestre varía sistemáticamente con la latitud y la longitud (aproximándose a un dipolo). Este componente de gran escala no está relacionado con las heterogeneidades locales del subsuelo que interesan a la geotecnia. Se utiliza un modelo del Campo Geomagnético Internacional (IGRF, *International Geomagnetic Reference Field*) o un ajuste polinómico de bajo orden a los datos de campo para eliminar esta tendencia regional, obteniendo así la **anomalía magnética residual**.

*   **Corrección por Deriva Instrumental:** Al igual que en la gravimetría, los magnetómetros pueden presentar una pequeña deriva con el tiempo debido a factores internos (temperatura, electrónica). Aunque generalmente es menor que la deriva gravimétrica, se controla y corrige con mediciones repetidas en la estación base o puntos de referencia.

*   **Eliminación de Ruido Cultural (Cultural Noise Removal):** Este es un paso crítico en magnetometría aplicada a geotecnia. Objetos metálicos (tuberías, cercas, vehículos, cables eléctricos, edificios) generan anomalías magnéticas muy intensas y de corta longitud de onda que pueden enmascarar las anomalías geológicas. Se requiere un cuidadoso control de campo para evitar la medición cerca de estas fuentes de ruido y, en el procesamiento, se pueden usar filtros espaciales (ej., filtros de alta frecuencia) o métodos de modelado para intentar aislar o eliminar estas señales de ruido.

Las **anomalías magnéticas residuales** resultantes son las que se interpretan en términos de contraste de susceptibilidad magnética en el subsuelo, permitiendo la identificación de objetos metálicos, ciertos tipos de rocas, o estructuras geológicas con mineralizaciones magnéticas.

La aplicación rigurosa de estas correcciones es fundamental para que los datos gravimétricos y magnéticos sean útiles en la caracterización geotécnica del subsuelo, permitiendo la detección y delimitación de anomalías que, de otro modo, permanecerían ocultas bajo la influencia de efectos más amplios.
