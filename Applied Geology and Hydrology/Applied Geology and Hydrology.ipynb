## 1.1. Tipología y clasificación de depósitos

### Tipología y clasificación de depósitos (Aplicado a Geotecnia)

La correcta identificación y clasificación de los depósitos geológicos es fundamental en geotecnia, ya que las propiedades ingenieriles de los suelos y rocas dependen en gran medida de su origen, composición y proceso de formación. Para una maestría en geotecnia, el enfoque no solo es descriptivo, sino que profundiza en las implicaciones que cada tipo de depósito tiene para el diseño y construcción de obras civiles.

#### Clasificación según el Origen y Proceso de Formación:

1.  **Depósitos Residuales (Suelos Residuales):**
    *   **Formación:** Resultado de la meteorización in situ de la roca madre. La roca se descompone química y/o físicamente sin que haya transporte significativo del material.
    *   **Características Geotécnicas:** Gran variabilidad en propiedades (desde arcillas blandas hasta suelos granulares densos o saprolitas). A menudo presentan una gradación gradual de material meteorizado a roca sana con la profundidad. Pueden contener texturas heredadas de la roca madre (relictas). La presencia de minerales arcillosos primarios (ej. caolinita en granitos meteorizados) es común.
    *   **Implicaciones en Geotecnia:** Dificultad en la caracterización debido a su heterogeneidad. Problemas de estabilidad de taludes, capacidad portante variable y potencial de colapso en suelos lixiviados o alterados. Se requiere una exploración geotécnica muy detallada.

2.  **Depósitos Transportados (Suelos Transportados):**
    *   **Formación:** Material que ha sido erosionado, transportado y depositado por agentes como el agua, el viento, el hielo o la gravedad.
    *   **Subtipos Clave en Geotecnia:**
        *   **Depósitos Aluviales:** Transportados y depositados por corrientes de agua (ríos, arroyos). Suelen ser arenas y gravas bien redondeadas, pero pueden incluir limos y arcillas en zonas de baja energía. Presentan estratificación y variabilidad lateral y vertical. **Implicaciones:** Buena capacidad portante en arenas y gravas densas, pero potencial de licuefacción en arenas sueltas saturadas. Problemas de cimentación por la heterogeneidad de los estratos y la presencia de lentes de materiales finos.
        *   **Depósitos Lacustres:** Depositados en lagos. Predominan limos y arcillas de baja plasticidad, a menudo laminados. **Implicaciones:** Suelos blandos y compresibles, con baja resistencia al corte. Problemas de asentamientos significativos y estabilidad de taludes. Alta sensibilidad al disturbio.
        *   **Depósitos Marinos:** Depositados en ambientes marinos. Incluyen arcillas sensibles (ej. arcillas de Leda en Canadá), limos y arenas. **Implicaciones:** Suelos muy blandos, altamente compresibles y con baja resistencia al corte, a menudo sobreeconsolidados por erosión. La presencia de sales puede afectar la durabilidad del concreto y acero.
        *   **Depósitos Eólicos (Loess, Dunas):** Transportados por el viento. El loess es limo uniforme, cohesivo y poroso; las dunas son arenas finas. **Implicaciones:** El loess es propenso a colapso por humedecimiento, muy compresible. Las arenas de dunas son sueltas y susceptibles a licuefacción y erosión eólica.
        *   **Depósitos Glaciales (Till, Outwash):** Transportados por glaciares. El till es una mezcla no estratificada de tamaños (desde arcilla hasta bloques); el outwash son arenas y gravas estratificadas por aguas de deshielo. **Implicaciones:** El till es generalmente denso y con buena capacidad portante, pero muy heterogéneo. El outwash tiene propiedades similares a los aluviales, pero con menor redondez.
        *   **Depósitos Coluviales:** Transportados por gravedad (desprendimientos, flujos de detritos). Son una mezcla heterogénea de bloques de roca, grava, arena y finos, a menudo angulosos. **Implicaciones:** Suelos muy variables, con potencial de estabilidad reducido en taludes, debido a la falta de compactación y la presencia de bloques. Alta permeabilidad y susceptibilidad a flujos de lodo.
        *   **Depósitos Volcánicos:** Incluyen cenizas, lapilli, tobas y rocas piroclásticas. **Implicaciones:** Pueden formar suelos cohesivos (arcillas alofánicas) o granulares. Algunos depósitos de ceniza son muy compresibles y susceptibles al colapso. Las tobas pueden ser muy alterables.

#### Importancia en Geotecnia:

*   **Estudios de Sitio:** La tipología de los depósitos guía la estrategia de exploración geotécnica (ubicación y tipo de sondeos, ensayos).
*   **Caracterización de Materiales:** Cada tipo de depósito tiene un rango esperado de propiedades geotécnicas (resistencia, compresibilidad, permeabilidad, plasticidad, etc.).
*   **Análisis y Diseño:** La comprensión del origen ayuda a prever el comportamiento de los suelos bajo carga y a diseñar cimentaciones, taludes, terraplenes y otras estructuras de manera segura y económica.
*   **Evaluación de Riesgos:** Asociados a fenómenos como deslizamientos, asentamientos, licuefacción o erosión, los cuales están directamente relacionados con el tipo de depósito presente.

En resumen, una clasificación geotécnica de depósitos va más allá de la geología descriptiva, enfocándose en cómo el origen y la formación impactan directamente en las propiedades ingenieriles y, por ende, en la interacción suelo-estructura y la estabilidad de las obras.

## 1.2. Ambientes sedimentarios y sus principales características

### Ambientes sedimentarios y sus principales características (Aplicado a Geotecnia)

La comprensión de los ambientes sedimentarios es crucial en geotecnia, ya que el ambiente de depósito controla la litología, la granulometría, la forma de los granos, la estratificación, la compacidad, el contenido de agua y la mineralogía de los suelos. Estos factores, a su vez, determinan las propiedades ingenieriles de los materiales y su comportamiento bajo diferentes condiciones de carga y estrés.

#### Concepto de Ambiente Sedimentario:

Un ambiente sedimentario es una zona geográfica donde los procesos físicos, químicos y biológicos operan de manera constante para producir un tipo característico de depósito sedimentario. Estos ambientes pueden ser clasificados en terrestres, de transición y marinos.

#### 1. Ambientes Terrestres (Continentales):

Engloban todos los depósitos formados en tierra firme, lejos de la influencia mareal y fluvial directa.

*   **Fluviales (Ríos y Arroyos):**
    *   **Características Geológicas:** Depósitos de arena y grava en los cauces (barras, lechos), y limos y arcillas en las llanuras de inundación (sobrebancos). Suelen ser heterogéneos lateral y verticalmente. Se observan estructuras como estratificación cruzada.
    *   **Implicaciones Geotécnicas:** Las arenas y gravas son materiales con buena capacidad portante si están densas, pero las arenas sueltas pueden licuarse. Los limos y arcillas de llanuras de inundación son blandos, compresibles y con baja resistencia al corte, propensos a asentamientos. La variabilidad del subsuelo fluvial es un desafío para la cimentación.

*   **Lacustres (Lagos):**
    *   **Características Geológicas:** Depósitos de grano fino (limos y arcillas) en el centro del lago, con arenas y gravas en las orillas. Frecuente laminación y alto contenido orgánico en algunos casos.
    *   **Implicaciones Geotécnicas:** Materiales de baja resistencia y alta compresibilidad, lo que genera problemas de estabilidad y grandes asentamientos. La presencia de materia orgánica puede generar gases y reducir la resistencia. Puede haber arcillas sensitivas.

*   **Glaciales:**
    *   **Características Geológicas:**
        *   **Till (morrena):** Depósito no clasificado, heterogéneo, desde arcilla hasta bloques, depositado directamente por el hielo.
        *   **Outwash:** Arenas y gravas estratificadas, depositadas por aguas de deshielo.
    *   **Implicaciones Geotécnicas:** El till puede ser muy denso y resistente, pero su heterogeneidad dificulta la caracterización. El outwash tiene propiedades similares a los aluviales, con buena capacidad portante en estado denso.

*   **Eólicos (Desiertos, Zonas Periglaciares):**
    *   **Características Geológicas:** Arenas bien clasificadas (dunas) y limos uniformes (loess). Estructuras de estratificación cruzada a gran escala en dunas.
    *   **Implicaciones Geotécnicas:** Las arenas eólicas son sueltas, con riesgo de licuefacción y erosión. El loess es cohesivo, pero altamente poroso y propenso a colapso por humedecimiento, causando asentamientos repentinos y significativos.

*   **Coluviales (Laderas):**
    *   **Características Geológicas:** Depósitos heterogéneos de detritos angulosos transportados por gravedad (deslizamientos, taludes). Suelen ser mezclas de grava, arena, limo y arcilla.
    *   **Implicaciones Geotécnicas:** Gran variabilidad en compacidad y resistencia. Riesgo de estabilidad de taludes, ya que son materiales a menudo no consolidados y susceptibles a movimientos. Exploración difícil debido a la presencia de bloques.

#### 2. Ambientes de Transición:

Zonas entre ambientes terrestres y marinos, influenciadas por ambos.

*   **Deltas:**
    *   **Características Geológicas:** Mezcla compleja de arenas, limos y arcillas, con rápida variación lateral y vertical debido a la interacción fluvial y marina. Rápida acumulación de sedimentos.
    *   **Implicaciones Geotécnicas:** Suelos altamente compresibles y blandos (especialmente los finos), con riesgo de asentamientos significativos y consolidación primaria prolongada. Inestabilidad por licuefacción de arenas en zonas sísmicas.

*   **Estuarios:**
    *   **Características Geológicas:** Valles fluviales inundados por el mar, con mezcla de agua dulce y salada. Predominan limos y arcillas con alto contenido orgánico. Pueden presentar "arcillas sensibles".
    *   **Implicaciones Geotécnicas:** Suelos blandos a muy blandos, altamente compresibles, con baja resistencia al corte y alta sensitividad. Problemas de asentamientos y estabilidad. La salinidad puede afectar la corrosión de las estructuras.

*   **Playas y Barreras Litorales:**
    *   **Características Geológicas:** Arenas y gravas bien clasificadas, con buena compacidad por la acción del oleaje. Pueden encontrarse depósitos de conchas.
    *   **Implicaciones Geotécnicas:** Materiales generalmente con buena capacidad portante. Riesgo de erosión costera y licuefacción si las arenas son finas y sueltas.

#### 3. Ambientes Marinos:

Depósitos formados en el mar, desde la costa hasta las profundidades oceánicas.

*   **Plataforma Continental (Poco Profundo):**
    *   **Características Geológicas:** Arenas en zonas de alta energía (cerca de la costa), y limos y arcillas a mayor profundidad. Puede haber depósitos de conchas y corales.
    *   **Implicaciones Geotécnicas:** Propiedades variables. Las arcillas y limos pueden ser blandos y compresibles, similares a los lacustres. Importante considerar la acción del oleaje y las corrientes marinas.

*   **Talud y Cuenca Oceánica (Profundo):**
    *   **Características Geológicas:** Dominan arcillas, limos y sedimentos biogénicos (ej. lodos calcáreos, radioláreos). Ritmos de sedimentación muy lentos.
    *   **Implicaciones Geotécnicas:** Suelos muy blandos, altamente compresibles y de muy baja resistencia, aunque pueden estar sobreeconsolidados debido a la remoción de carga (erosión). Deslizamientos submarinos son un riesgo importante en el talud. La exploración y caracterización son complejas debido a la profundidad.

#### Consideraciones Geotécnicas Clave:

*   **Heterogeneidad:** La mayoría de los ambientes sedimentarios, especialmente los fluviales y deltaicos, generan depósitos con gran variabilidad. Esto requiere una exploración geotécnica intensiva.
*   **Contenido de Agua y Compresibilidad:** Los depósitos finos (limos y arcillas) de ambientes lacustres, marinos y de llanura de inundación suelen tener alto contenido de agua, baja resistencia y alta compresibilidad, lo que lleva a grandes asentamientos.
*   **Potencial de Licuefacción:** Las arenas sueltas y saturadas en ambientes fluviales, eólicos y de playa son susceptibles a la licuefacción bajo carga sísmica.
*   **Mineralogía de Arcillas:** La mineralogía de las arcillas (ej. caolinita, ilita, esmectita) está influenciada por el ambiente de formación y afecta directamente la plasticidad, compresibilidad e hinchamiento de los suelos.

En una maestría de geotecnia, el estudio de los ambientes sedimentarios no solo busca describir, sino comprender los procesos que llevan a la formación de suelos con ciertas características ingenieriles, permitiendo así una mejor predicción de su comportamiento y un diseño más seguro y eficiente de las obras civiles.

# 2. Estructuras geológicas y tectónicas a escala regional y local

## 2.1. Falla, pliegues y diaclasas

### Falla, pliegues y diaclasas (Aplicado a Geotecnia)

Las estructuras geológicas, como fallas, pliegues y diaclasas, son de vital importancia en geotecnia, ya que controlan significativamente el comportamiento mecánico de los macizos rocosos y de los suelos adyacentes. Su correcta identificación, caracterización y análisis son fundamentales para el diseño seguro y económico de cimentaciones, túneles, taludes y otras estructuras de ingeniería civil.

#### 1. Fallas:

*   **Definición:** Son fracturas en la corteza terrestre a lo largo de las cuales ha ocurrido un desplazamiento relativo de los bloques de roca.
*   **Tipos Principales y su Relevancia Geotécnica:**
    *   **Fallas Normales (Extensionales):** Resultan de fuerzas tensionales. Pueden generar zonas de roca muy fragmentada en el plano de falla, lo que implica baja resistencia al corte y alta permeabilidad. Riesgo de inestabilidad de taludes y subsidencia.
    *   **Fallas Inversas y Cabalgamientos (Compresionales):** Resultan de fuerzas compresionales. Crean zonas de roca intensamente deformada y cizallada, con cambios abruptos en la litología y la calidad de la roca. Pueden ser planos de debilidad críticos para la estabilidad de túneles y taludes. La superposición de unidades por cabalgamiento puede generar anomalías en la secuencia estratigráfica y en las propiedades del subsuelo.
    *   **Fallas de Desgarre o Rumbo (Transcurrentes):** El movimiento es predominantemente horizontal. Generan zonas de cizalla extensas (zonas de cizalla), con roca intensamente triturada (brechas de falla, harinas de falla o gouge), que tienen muy baja resistencia y alta compresibilidad. Esto afecta gravemente la capacidad portante y la estabilidad. Son críticas en zonas sísmicas por el riesgo de ruptura superficial.
*   **Implicaciones Geotécnicas Clave:**
    *   **Zonas de Debilidad:** Los planos de falla y las zonas de influencia adyacentes son intrínsecamente débiles debido a la trituración y alteración de la roca. La resistencia al corte de estos materiales es mucho menor que la de la roca intacta.
    *   **Permeabilidad:** Las fallas actúan como conductos o barreras para el flujo de agua subterránea, dependiendo de la naturaleza del material de relleno. Esto es crucial en proyectos de túneles y presas.
    *   **Sismicidad:** Las fallas activas son fuentes de terremotos. La proximidad a fallas activas impone requisitos de diseño sísmico rigurosos para las estructuras.
    *   **Asentamientos Diferenciales:** Los cambios bruscos en las propiedades del suelo/roca a través de una falla pueden causar asentamientos diferenciales en estructuras cimentadas sobre ellas.

#### 2. Pliegues:

*   **Definición:** Son deformaciones dúctiles de las capas de roca que resultan en curvaturas. Los tipos principales son anticlinales (arcos hacia arriba) y sinclinales (arcos hacia abajo).
*   **Implicaciones Geotécnicas Clave:**
    *   **Variación de la Orientación de los Planos de Discontinuidad:** En un flanco de pliegue, los planos de estratificación y otras discontinuidades (diaclasas) tienen una orientación diferente a la del otro flanco o a la de la zona de charnela. Esto afecta la estabilidad de taludes y túneles, ya que la resistencia al corte es anisotrópica.
    *   **Zonas de Charnela (Cresta y Valle):** En las charnelas de los pliegues (especialmente en pliegues apretados), las rocas pueden estar más fracturadas (diaclasamiento intenso), lo que reduce su resistencia y aumenta su permeabilidad. Esto es crítico al excavar túneles a través de estas zonas.
    *   **Aumento de Esfuerzos:** Las zonas de pliegue pueden concentrar esfuerzos, lo que es relevante en el diseño de excavaciones subterráneas.
    *   **Trampas de Agua:** Las estructuras plegadas pueden crear trampas o acuíferos, especialmente los sinclinales, lo que influye en el régimen hidrológico de un área.

#### 3. Diaclasas (Juntas o Fracturas):

*   **Definición:** Son fracturas en la roca sin desplazamiento relativo significativo paralelo al plano de fractura (a diferencia de las fallas).
*   **Origen:** Pueden ser el resultado de esfuerzos tectónicos, liberación de carga, enfriamiento de rocas ígneas, contracción por desecación en rocas sedimentarias, etc.
*   **Implicaciones Geotécnicas Clave:**
    *   **Reducción de la Resistencia del Macizo Rocoso:** Las diaclasas dividen el macizo rocoso en bloques. La resistencia global del macizo está controlada por la resistencia al corte a lo largo de estos planos de discontinuidad, que es mucho menor que la de la roca intacta. Un sistema de diaclasas bien desarrollado puede transformar una roca competente en un material muy débil.
    *   **Estabilidad de Taludes y Túneles:** La orientación de los sistemas de diaclasas con respecto a la excavación es el factor más crítico. Discontinuidades desfavorablemente orientadas pueden causar fallas por cuña, falla planar o vuelco. La resistencia al corte de las diaclasas (cohesión y ángulo de fricción) es fundamental en los análisis de estabilidad.
    *   **Permeabilidad y Flujo de Agua:** Las diaclasas son las principales vías para el flujo de agua subterránea en macizos rocosos. Esto afecta la presión de poros, la estabilidad (al reducir la resistencia al corte), y puede generar problemas de entrada de agua en túneles o filtraciones en presas.
    *   **Anisotropía:** Un macizo rocoso con diaclasas exhibe propiedades anisotrópicas, lo que significa que su resistencia y deformabilidad varían con la dirección.
    *   **Volumen de Roca Excavada:** El espaciamiento y la orientación de las diaclasas influyen en el tamaño de los bloques que se obtienen durante la excavación, afectando la estabilidad y el método de voladura.

#### Metodologías de Caracterización Geotécnica:

Para todas estas estructuras, la geotecnia de maestría enfatiza:

*   **Cartografía Geomecánica:** Mapeo detallado de la orientación (rumbo y buzamiento), espaciamiento, persistencia, abertura, rugosidad y relleno de las discontinuidades.
*   **Estereografía:** Uso de la proyección estereográfica para analizar la orientación de las discontinuidades y evaluar la estabilidad de taludes y excavaciones subterráneas.
*   **Clasificaciones Geomecánicas:** Aplicación de sistemas como RMR (Rock Mass Rating) y Q-system para cuantificar la calidad del macizo rocoso en función de las propiedades de las discontinuidades.
*   **Ensayos de Laboratorio y Campo:** Medición de la resistencia al corte de las discontinuidades (ej. ensayo de cizalla directa).
*   **Modelado Numérico:** Simulación del comportamiento del macizo rocoso considerando la presencia y propiedades de fallas, pliegues y diaclasas.

## 2.2. Medida de rumbo y buzamiento

### Medida de rumbo y buzamiento (Aplicado a Geotecnia)

La medida del rumbo y el buzamiento es una de las operaciones fundamentales en geología estructural y geotecnia. Permite caracterizar la orientación espacial de los planos de discontinuidad (estratificación, fallas, diaclasas, foliación, etc.) que afectan a un macizo rocoso o a un cuerpo de suelo. Esta información es crítica para el análisis de estabilidad y el diseño geotécnico de obras de ingeniería civil.

#### Conceptos Fundamentales:

1.  **Rumbo (Strike):**
    *   **Definición:** Es la dirección de la línea de intersección entre un plano geológico (por ejemplo, un estrato, una falla o una diaclasa) y un plano horizontal. Se mide como un ángulo con respecto al Norte geográfico (0° o 360°), en sentido horario, y se expresa generalmente entre 0° y 360°.
    *   **Notación:** Por ejemplo, N45°E o 045°. También puede expresarse indicando la dirección de la línea (ej., 045/225, donde 045 es la dirección de menor ángulo y 225 su opuesta).
    *   **Importancia Geotécnica:** Define la dirección general del plano en planta. Un talud paralelo al rumbo de un plano de discontinuidad puede ser susceptible a fallas planares si el buzamiento es hacia el talud. Es fundamental para orientar los mapas geológicos y los perfiles geotécnicos.

2.  **Buzamiento (Dip):**
    *   **Definición:** Es el ángulo de máxima pendiente de un plano geológico con respecto a la horizontal. Se mide en un plano vertical perpendicular al rumbo, y su valor oscila entre 0° (horizontal) y 90° (vertical). Siempre se debe indicar también la dirección del buzamiento.
    *   **Dirección del Buzamiento (Dip Direction):** Es la dirección horizontal hacia la cual el plano inclina. Se mide también en sentido horario desde el Norte geográfico y se expresa entre 0° y 360°.
    *   **Notación:** Se expresa como ángulo de buzamiento/dirección de buzamiento (ej., 30°/135°). Alternativamente, puede indicarse el rumbo y luego el buzamiento con la dirección cardinal (ej., N45°E, 30°SE).
    *   **Importancia Geotécnica:** El ángulo y la dirección del buzamiento son los parámetros más críticos. Un plano que buza hacia un talud con un ángulo similar al del talud es una condición muy desfavorable para la estabilidad. El buzamiento controla la facilidad con la que el agua puede infiltrarse en el macizo rocoso y las direcciones preferenciales de flujo.

#### Herramientas de Medida:

*   **Brújula Geológica:** La herramienta principal. Las más comunes son la Brújula Brunton y la Brújula de Clar.
    *   **Brújula Brunton:** Mide directamente el rumbo y el buzamiento/dirección de buzamiento. Requiere práctica para asegurar la precisión.
    *   **Brújula de Clar (o Suunto/Freiberg):** Mide el rumbo como acimut y el buzamiento con una aguja y escala lateral. A menudo se utiliza para medir la dirección del buzamiento y el ángulo de buzamiento directamente. Es importante recordar que la lectura del rumbo es perpendicular a la dirección del buzamiento.
*   **Clinómetro/Nivel de Burbuja:** Para verificar la horizontalidad de la brújula y medir ángulos de inclinación. Algunos teléfonos inteligentes y aplicaciones específicas también pueden ser útiles como clinómetros.
*   **Software de Posprocesamiento:** Programas como DIPS (Rocscience), Stereonet o Spheristat para la representación gráfica y el análisis estadístico de los datos.

#### Procedimiento Básico de Medida (con Brújula de Clar/Suunto):

1.  **Medida de Rumbo:**
    *   Colocar la brújula sobre el plano de discontinuidad de manera que el borde largo de la brújula esté en contacto con el plano y lo más horizontal posible.
    *   Rotar la brújula hasta que el nivel de burbuja esté centrado (indicando horizontalidad). La aguja de la brújula indicará la dirección del rumbo. Anotar el valor.
    *   Es fundamental asegurarse de que se está midiendo la línea horizontal del plano.

2.  **Medida de Buzamiento:**
    *   Girar la brújula 90° respecto al rumbo, de modo que el borde largo esté todavía en contacto con el plano, pero ahora a lo largo de la línea de máxima pendiente.
    *   Usar el clinómetro para medir el ángulo de inclinación del plano. Anotar el valor del ángulo de buzamiento.
    *   Determinar la dirección de buzamiento observando hacia dónde inclina el plano (ej., N, NE, E, SE, S, SW, W, NW). Para una mayor precisión, se puede medir la dirección con la brújula apuntando cuesta abajo en la línea de máxima pendiente del plano.

#### Consideraciones Cruciales en Geotecnia:

*   **Precisión:** La exactitud en las mediciones es vital, ya que pequeños errores pueden tener grandes implicaciones en el análisis de estabilidad.
*   **Representatividad:** Es necesario tomar suficientes medidas para caracterizar adecuadamente los diferentes sistemas de discontinuidades presentes en un macizo rocoso. No basta con unas pocas medidas aisladas.
*   **Variabilidad:** Reconocer que las orientaciones pueden variar ligeramente dentro de un mismo sistema debido a la heterogeneidad del macizo o a deformaciones menores. Esto se aborda con análisis estadísticos en redes estereográficas.
*   **Efectos Locales:** Tener en cuenta si hay efectos locales (por ejemplo, bloques inestables, erosión diferencial) que puedan alterar la orientación de un plano.
*   **Unidades de Medida:** Mantener la consistencia en el sistema de coordenadas y las convenciones de rumbo/buzamiento para evitar confusiones.
*   **Impacto del Agua:** Las discontinuidades son las principales vías de flujo de agua. Su orientación controla la dirección del flujo preferencial y la presión de poros, factores clave en la estabilidad de taludes y túneles.

En resumen, la medida de rumbo y buzamiento es una habilidad básica pero poderosa que permite a los ingenieros geotécnicos transformar la geometría compleja del subsuelo en datos cuantificables para su análisis y diseño. Es el primer paso para entender el comportamiento cinemático de los macizos rocosos.
